# Part A - CNN fundamentals

## 3. Basic CNN operations as explicit functions

Before using TensorFlow or PyTorch, this section makes the essential CNN mathematics visible with small NumPy functions. This follows the conceptual direction of Lecture 04: first understand the computation, then express the same computation through a framework abstraction.

### 3.1 Neuron and linear transformation

A single affine transformation can be written as:

\[
z = w^T x + b
\]

where `x` is the input vector, `w` contains learnable weights, and `b` is a learnable bias. A CNN eventually uses these learned representations for classification, even though the earlier feature-extraction stages preserve spatial structure.

### 3.2 ReLU

ReLU introduces non-linearity:

\[
ReLU(x) = max(0, x)
\]

Without a non-linear activation, stacking linear layers would still produce a linear transformation. ReLU therefore allows successive layers to build progressively more complex representations.

### 3.3 Softmax and cross-entropy

For multiclass classification, logits are converted into probabilities with softmax. Cross-entropy then measures how much probability the model assigned to the true class.

The implementation below is intentionally small. Its purpose is not to replace a production framework, but to show the computation that framework layers ultimately perform.




````python
# Purpose: Demonstrate the basic numerical operations that appear inside a classifier.
# Each function below is intentionally small so that its mathematical role is visible.
def relu(x):
    """Apply the Rectified Linear Unit activation element by element.

    Purpose:
        Introduce non-linearity by keeping positive activations and replacing negative
        activations with zero.

    Parameters:
        x (np.ndarray): Input activation values.

    Returns:
        np.ndarray: ReLU-transformed values with the same shape as the input.
    """
    # ReLU keeps positive values and clips negative values to zero.
    return np.maximum(0, x)


def softmax(logits):
    """Convert multiclass logits into a normalized probability distribution.

    Purpose:
        Turn arbitrary class scores into probabilities whose class dimension sums to one.

    Parameters:
        logits (np.ndarray): Model scores with classes on the last axis.

    Returns:
        np.ndarray: Numerically stable class probabilities.
    """
    # Subtract the largest logit before exponentiation to avoid overflow.
    shifted = logits - np.max(logits, axis=-1, keepdims=True)
    exp_x = np.exp(shifted)
    return exp_x / exp_x.sum(axis=-1, keepdims=True)


def sparse_cross_entropy(logits, y_true):
    """Compute mean sparse multiclass cross-entropy.

    Purpose:
        Quantify the negative log-probability assigned to the correct class.

    Parameters:
        logits (np.ndarray): Unnormalized class scores with shape (N, C).
        y_true (np.ndarray): Integer class labels with shape (N,).

    Returns:
        float: Mean cross-entropy across the supplied samples.
    """
    # Convert logits to probabilities and select the probability of each true class.
    probs = softmax(logits)
    p_true = probs[np.arange(len(y_true)), y_true]
    # Clip probabilities so log() remains numerically safe for very small values.
    return float(-np.log(np.clip(p_true, 1e-12, 1.0)).mean())


# Purpose: Run a tiny numerical example so the functions can be inspected immediately.
values = np.array([-2.0, -1.0, 0.0, 1.0, 2.0])
logits = np.array([[2.0, 1.0, 0.1]])
y_true = np.array([0])

print("ReLU:", relu(values))
print("Softmax:", softmax(logits))
print("Cross-entropy:", sparse_cross_entropy(logits, y_true))


````

### 3.0 Why these functions matter

The point of the following functions is not to recreate a production deep-learning framework. Their role is pedagogical: they expose the mathematical operations that are later wrapped by high-level TensorFlow and PyTorch APIs.

When reading a framework model later, the reader should be able to map the layer names back to these elementary transformations. `Conv2D`/`Conv1D` performs learned local aggregation, `ReLU` supplies non-linearity, pooling or global pooling changes representation size, and the final linear layer converts learned features into class logits.



````python
# Purpose: Demonstrate spatial operations that form the core of a CNN feature extractor.
def conv2d_single_channel(image, kernel, stride=1, padding=0):
    """Perform a simple single-channel 2-D cross-correlation operation.

    Purpose:
        Make the local convolution computation visible without hiding it inside a framework layer.

    Parameters:
        image (np.ndarray): Two-dimensional input feature map.
        kernel (np.ndarray): Two-dimensional learnable filter.
        stride (int): Step size between neighboring windows.
        padding (int): Number of zero-padding cells added around the input.

    Returns:
        np.ndarray: Two-dimensional output feature map.
    """
    # Convert inputs to float32 so the example behaves like a neural-network tensor.
    image = np.asarray(image, dtype=np.float32)
    kernel = np.asarray(kernel, dtype=np.float32)

    # Padding preserves border information when the architecture requests it.
    if padding:
        image = np.pad(image, ((padding, padding), (padding, padding)))

    kh, kw = kernel.shape
    out_h = (image.shape[0] - kh) // stride + 1
    out_w = (image.shape[1] - kw) // stride + 1
    output = np.zeros((out_h, out_w), dtype=np.float32)

    for i in range(out_h):
        for j in range(out_w):
            # Each output value is a local element-wise product followed by a sum.
            region = image[i * stride:i * stride + kh, j * stride:j * stride + kw]
            output[i, j] = np.sum(region * kernel)

    return output


def max_pool2d(feature_map, pool_size=2, stride=2):
    """Apply non-overlapping 2-D max pooling.

    Purpose:
        Reduce spatial resolution while retaining the strongest local activation in each window.

    Parameters:
        feature_map (np.ndarray): Two-dimensional activation map.
        pool_size (int): Height and width of the pooling window.
        stride (int): Step size used when moving the pooling window.

    Returns:
        np.ndarray: Downsampled feature map containing local maxima.
    """
    h, w = feature_map.shape
    out_h = (h - pool_size) // stride + 1
    out_w = (w - pool_size) // stride + 1
    output = np.zeros((out_h, out_w), dtype=np.float32)

    for i in range(out_h):
        for j in range(out_w):
            # Pooling keeps the largest activation rather than computing a weighted sum.
            region = feature_map[
                i * stride:i * stride + pool_size,
                j * stride:j * stride + pool_size,
            ]
            output[i, j] = np.max(region)

    return output


# Purpose: Verify the spatial functions on a tiny hand-checkable example.
example_image = np.arange(25, dtype=np.float32).reshape(5, 5)
example_kernel = np.ones((3, 3), dtype=np.float32)

print("Convolution result:\n", conv2d_single_channel(example_image, example_kernel))
print("Pooling result:\n", max_pool2d(example_image))


````

## 3.1 Why compare CNN3 and CNN5?

In this report, **CNN3** and **CNN5** refer specifically to the number of convolutional layers. Batch Normalization, ReLU, pooling, global average pooling, dropout, and the final classification layer are not counted as convolutional depth.

This definition is important because otherwise two readers could count the same model differently.

### CNN3 design

```text
Input
  ↓
Conv 32 → BatchNorm → ReLU → Pool
  ↓
Conv 64 → BatchNorm → ReLU → Pool
  ↓
Conv 128 → BatchNorm → ReLU
  ↓
Global Average Pooling
  ↓
Dropout(0.30)
  ↓
Dense / Linear classifier
```

### CNN5 design

```text
Input
  ↓
Conv 32 → BatchNorm → ReLU → Pool
  ↓
Conv 64 → BatchNorm → ReLU → Pool
  ↓
Conv 128 → BatchNorm → ReLU
  ↓
Conv 128 → BatchNorm → ReLU → Pool
  ↓
Conv 160 → BatchNorm → ReLU
  ↓
Global Average Pooling
  ↓
Dropout(0.30)
  ↓
Dense / Linear classifier
```

### Why this is a useful controlled experiment

The first two convolutional stages are shared conceptually by both depths. CNN5 then adds two more convolutional transformations and one additional pooling stage. This increases representational capacity while keeping the basic family of operations unchanged.

A deeper model may improve representation quality, but depth also introduces more parameters, more computation, and more opportunities for optimization difficulty or overfitting. Therefore the report must compare accuracy together with validation behavior, macro F1, per-class recall, parameter count, and training time.

The experiment is not designed to assume that CNN5 will always win. The result must be established empirically for each dataset.




## 3.2 Shared audit, visualization, and evaluation utilities

A4 emphasized that model comparison should include more than a single accuracy value. A5 keeps that principle and expands the diagnostics because the three datasets have very different class structures.

The shared functions below are intentionally framework-independent. They receive predictions and probabilities, not TensorFlow or PyTorch model objects. This makes the metric definitions identical for both frameworks.

### Why centralize the evaluation logic?

Suppose TensorFlow uses one definition of F1 and PyTorch uses another, or one model is evaluated with probabilities while another is evaluated after a different thresholding step. The comparison would no longer be fair. Shared utilities reduce that risk.

The visualization suite includes:

1. class-count distributions;
2. image sample grids;
3. pixel/value distributions;
4. mean and variability maps;
5. training/validation loss and accuracy;
6. numeric confusion matrices;
7. per-class precision, recall, and F1;
8. confidence distributions;
9. reliability diagrams and expected calibration error;
10. misclassified image examples;
11. top confusion pairs;
12. unified experiment tables.

Each plot has a specific analytical purpose. The goal is to turn model output into interpretable evidence.




````python
# Purpose: Provide shared, framework-independent visualization and data-splitting utilities.
def plot_class_distribution(labels, class_names=None, title="Class distribution"):
    """Plot class counts with numeric annotations.

    Purpose:
        Inspect whether the target classes are balanced and provide exact counts for the report.

    Parameters:
        labels (np.ndarray): Integer class labels.
        class_names (list|None): Optional readable names for each class index.
        title (str): Figure title.

    Returns:
        None. The function displays the figure and does not modify the labels.
    """
    values, counts = np.unique(labels, return_counts=True)
    names = [class_names[int(v)] if class_names is not None else str(v) for v in values]
    plt.figure(figsize=(12, 5))
    bars = plt.bar(names, counts)
    plt.title(title)
    plt.xlabel("Class")
    plt.ylabel("Count")
    plt.xticks(rotation=45, ha="right")
    for bar, count in zip(bars, counts):
        # Write the exact sample count above every class for auditability.
        plt.text(bar.get_x()+bar.get_width()/2, bar.get_height(), f"{count:,}", ha="center", va="bottom", fontsize=8)
    plt.tight_layout(); plt.show()


def plot_image_grid(images, labels, class_names, n=12, title="Samples"):
    """Display a labeled grid of representative images.

    Purpose:
        Verify that image tensors and labels have been loaded and aligned correctly.

    Parameters:
        images (np.ndarray): Image array in HxW or HxWxC form.
        labels (np.ndarray): Corresponding integer labels.
        class_names (list): Human-readable class names.
        n (int): Maximum number of examples to display.
        title (str): Figure title.

    Returns:
        None. The plot is displayed for visual inspection.
    """
    n = min(n, len(images))
    cols = 4
    rows = math.ceil(n / cols)
    plt.figure(figsize=(12, 3*rows))
    for i in range(n):
        ax = plt.subplot(rows, cols, i+1)
        # Grayscale images use a gray colormap; RGB images keep their channels unchanged.
        ax.imshow(images[i], cmap="gray" if images[i].ndim == 2 else None)
        ax.set_title(class_names[int(labels[i])])
        ax.axis("off")
    plt.suptitle(title); plt.tight_layout(); plt.show()


def plot_image_statistics(images, title):
    """Visualize pixel-value distribution, mean image, and spatial variability.

    Purpose:
        Inspect intensity distribution and whether some image regions vary more than others.

    Parameters:
        images (np.ndarray): Raw image array.
        title (str): Dataset name used in figure titles.

    Returns:
        None. Three diagnostic views are displayed.
    """
    flat = images.astype(np.float32).reshape(-1)
    plt.figure(figsize=(9, 4))
    plt.hist(flat, bins=50)
    plt.title(title + " - pixel distribution")
    plt.xlabel("Pixel value"); plt.ylabel("Frequency")
    plt.tight_layout(); plt.show()

    mean_img = images.astype(np.float32).mean(axis=0)
    std_img = images.astype(np.float32).std(axis=0)
    if mean_img.ndim == 3:
        # Average RGB channels only for visualization of spatial structure.
        mean_img = mean_img.mean(axis=-1)
        std_img = std_img.mean(axis=-1)
    fig, ax = plt.subplots(1, 2, figsize=(10, 4))
    ax[0].imshow(mean_img, cmap="gray"); ax[0].set_title("Mean image"); ax[0].axis("off")
    ax[1].imshow(std_img, cmap="magma"); ax[1].set_title("Pixel std map"); ax[1].axis("off")
    plt.suptitle(title); plt.tight_layout(); plt.show()


def report_image_integrity(images, labels, name, n_classes):
    """Run basic integrity checks before image preprocessing.

    Purpose:
        Detect malformed values, invalid labels, and unexpected ranges before training begins.

    Parameters:
        images (np.ndarray): Raw image array.
        labels (np.ndarray): Integer class labels.
        name (str): Dataset display name.
        n_classes (int): Expected number of classes.

    Returns:
        None. Audit results are printed for the report.
    """
    x = images.astype(np.float32)
    print(f"[{name}]")
    print("shape:", images.shape, "dtype:", images.dtype)
    print("NaN:", int(np.isnan(x).sum()), "Inf:", int(np.isinf(x).sum()))
    print("pixel range:", float(x.min()), "to", float(x.max()))
    print("invalid labels:", int(((labels < 0) | (labels >= n_classes)).sum()))


def split_image_train_val(X, y, val_size=0.10, seed=42, train_limit=None, val_limit=None):
    """Create a stratified train/validation split and optional deterministic pilot limits.

    Purpose:
        Keep class proportions comparable while allowing a smaller development run in pilot mode.

    Parameters:
        X (np.ndarray): Image samples.
        y (np.ndarray): Integer labels.
        val_size (float): Fraction assigned to validation before optional limiting.
        seed (int): Split seed.
        train_limit (int|None): Maximum training rows for pilot mode.
        val_limit (int|None): Maximum validation rows for pilot mode.

    Returns:
        tuple[np.ndarray, np.ndarray]: Sorted training and validation indices.
    """
    idx = np.arange(len(y))
    # Stratification makes the framework comparison use similar class proportions.
    train_idx, val_idx = train_test_split(idx, test_size=val_size, stratify=y, random_state=seed)
    if train_limit is not None and train_limit < len(train_idx):
        train_idx, _ = train_test_split(train_idx, train_size=train_limit, stratify=y[train_idx], random_state=seed)
    if val_limit is not None and val_limit < len(val_idx):
        val_idx, _ = train_test_split(val_idx, train_size=val_limit, stratify=y[val_idx], random_state=seed)
    return np.sort(train_idx), np.sort(val_idx)


````

### 4.0 Raw representation versus model representation

The raw Fashion-MNIST images are stored as integer pixel values. The model does not need those values to remain integers. Neural networks generally operate on floating-point tensors so gradient-based optimization can update their parameters smoothly.

The transformation therefore has two distinct purposes:

1. preserve the original raw data for auditability;
2. create a normalized floating-point representation for optimization.

Keeping those stages conceptually separate is useful when debugging a web demo later: the uploaded image is raw input, while the model receives the exact preprocessed tensor used during training.



## 4.2 What the Fashion-MNIST audit means

A curated benchmark does not require aggressive cleaning. In a well-designed machine-learning pipeline, "cleaning" should mean verifying the data before transforming it, not deleting samples just because a model later finds them difficult.

The audit therefore checks:

- image and label counts;
- data shape and dtype;
- minimum and maximum pixel values;
- NaN and infinite values;
- whether labels are inside the expected class range;
- duplicate rows;
- class counts.

For Fashion-MNIST, the main modeling transformations are numerical rather than corrective:

```text
uint8 pixels 0-255
      ↓
float32
      ↓
normalization to 0-1
      ↓
training-only augmentation
      ↓
CNN input
```

Normalizing pixel values makes optimization numerically better behaved. The conversion is deterministic and does not use any statistics from the test set.

The official test set is kept untouched until the final evaluation stage. This separation protects the test set from becoming an informal validation set during experimentation.




## 4.1 Cleaning, integrity checks, preprocessing, and split

For the model comparison, the original training partition is divided into training and validation data using a stratified split. Stratification preserves approximately the same class proportions in both partitions.

The test partition remains the official Fashion-MNIST test set.

### Why validation is necessary

The training set is used to update model parameters. The validation set is used to make development decisions such as:

- whether the model is still improving;
- whether early stopping should occur;
- which checkpoint should be retained.

The test set should not influence those choices. It is reserved for the final performance estimate after the model-selection process is complete.

### Pilot mode

In pilot mode, the same splitting procedure is used but only a deterministic subset of the training and validation partitions is retained. The test set is not replaced by a smaller arbitrary slice for the final pipeline; the intent is to keep the debugging path representative while avoiding unnecessary compute.

### 5.0 CIFAR-100 exploratory analysis

For a 100-class image problem, sample visualization should be intentionally diverse. The reader should see examples from multiple regions of the label space instead of a grid containing only one or two classes.

The class-count plot also serves as an integrity check. A balanced benchmark should show approximately equal counts across fine classes. A surprising deviation is a reason to investigate the loading pipeline before training.



## 5.2 Why CIFAR-100 requires richer evaluation

Accuracy remains important, but it is only one view of a 100-class classifier.

### Top-1 accuracy

The predicted class is the single class with maximum probability. Top-1 accuracy asks whether that class matches the target.

### Top-5 accuracy

The five highest-probability classes are considered. A test sample counts as correct if the true class appears anywhere inside those five candidates. This metric is especially informative for large class spaces because it measures whether the model has learned a useful ranking even when the top decision is incorrect.

### Macro F1

Macro F1 calculates F1 for each class and then gives each class equal weight. This prevents the most frequent classes from completely determining the summary score.

### Confusion structure

With 100 classes, the full 100 x 100 matrix is retained as evidence, but a second readable summary lists the strongest off-diagonal confusion pairs. This allows the report to answer questions such as "Which classes are repeatedly mixed together?" rather than only "How many total mistakes were made?"




## 5.1 Cleaning, integrity checks, preprocessing, and split

CIFAR-100 is already distributed as a balanced benchmark, so the preprocessing objective is primarily integrity preservation and numerical preparation.

The audit verifies that:

- every image has shape 32 x 32 x 3;
- the labels cover the expected 100 classes;
- values are finite and remain in the expected raw range;
- the number of images equals the number of labels;
- class counts are consistent with the official distribution.

The training set is then split into train and validation partitions using stratification. The official test set remains isolated for final evaluation.

### Normalization

RGB channels are converted to floating point. Normalization is performed according to the training data statistics used by the pipeline. The important methodological rule is that any learned preprocessing statistics must be estimated from the training partition only.

### Augmentation

The training pipeline uses a light horizontal-flip augmentation. Augmentation is applied only during training. Validation and test images remain unchanged so that the evaluation answers the same question for every framework.

### 6.0.1 Interpreting missingness

A missingness table answers two questions before any imputation is performed: **which variables are incomplete, and how incomplete are they?**

A high missingness percentage does not automatically prove that a variable must be deleted. The decision depends on how the variable is represented, whether its values are usable at inference time, and whether its absence is itself meaningful. In this assignment, the notebook documents explicit drop decisions and makes the remaining imputation rules visible.



## 6.3 Patient leakage: why a random row split is not sufficient

The dataset contains multiple encounters for the same patient. That means the rows are not necessarily independent observations at the patient level.

Consider a patient with several hospital encounters. If some encounters are placed in the training set while another encounter from the same patient is placed in the test set, the model can benefit from patient-specific patterns that were already present during training. The resulting test set is less independent than it appears.

### Group-aware splitting principle

The notebook therefore treats `patient_nbr` as a grouping variable during partitioning:

```text
Patient IDs
    ↓
Group-aware partitioning
    ↓
Training patients      Validation patients      Test patients
        |                       |                     |
     disjoint               disjoint              disjoint
```

After the split, the notebook explicitly computes the intersection size between the patient ID sets. A correctly separated experiment should report zero overlap for every partition pair.

This is one of the most important methodological controls in the A5 pipeline because it concerns the independence of the evaluation data, not just the appearance of the score.




## 6.1 Cleaning strategy for the clinical table

The cleaning process is documented as a sequence of decisions rather than hidden inside one line of code.

### Step 1 - Missing-value representation

The original data use `?` as a sentinel for missing information. That sentinel is replaced with `NaN` so that Pandas and scikit-learn can measure and process missingness explicitly.

### Step 2 - Target validation

Rows without a valid target are not useful for supervised training and are removed before model construction.

### Step 3 - Identifier handling

`encounter_id` identifies the encounter and `patient_nbr` identifies the patient. They are not treated as ordinary predictive features because they can encode identity rather than clinically meaningful signal.

`patient_nbr` is temporarily retained for the group-aware split and removed from the model input afterwards.

### Step 4 - Highly missing or unsuitable variables

Variables with extremely high missingness are documented as explicit drop decisions rather than being silently passed into the model. The notebook records which columns are dropped so that the preprocessing is auditable.

### Step 5 - Numeric and categorical preprocessing

Numerical columns use median imputation followed by standardization. Categorical columns use most-frequent imputation, ordinal encoding with an explicit unknown value, and standardization.

The ordinal encoding is a compact representation chosen for the Conv1D experiment. It does **not** claim that category code 3 is intrinsically three times or twice as meaningful as category code 1. The code represents category membership so that the network can consume a fixed-length tensor.

### Step 6 - Training-only fitting

The preprocessing pipeline is fitted using training rows only. Validation and test rows are transformed using the already-fitted pipeline.

This rule is fundamental:

```text
TRAIN: fit preprocessing + transform
VAL  : transform only
TEST : transform only
```

Fitting a scaler or imputer on the complete dataset would allow information from validation/test data to influence the representation used during training.




````python
# Purpose: Run the computation defined by this section. The detailed comments below explain the role of each operation.
# Original section purpose is described in the preceding Markdown cell. 26.
def prepare_diabetes(df, seed=42, run_mode="full"):
    """Clean, split, encode, scale, and reshape the Diabetes dataset for Conv1D.

    Purpose:
        Build one reproducible preprocessing pipeline while preventing patient-level and
        preprocessing leakage between train, validation, and test partitions.

    Parameters:
        df (pd.DataFrame): Raw UCI Diabetes dataframe containing features and `readmitted`.
        seed (int): Random seed for the group-aware split and pilot sampling.
        run_mode (str): "pilot" for development subsets or "full" for the intended dataset size.

    Returns:
        dict: Processed tensors, labels, preprocessing metadata, class weights, and leakage checks.
    """
    # Convert the source missing-value sentinel into a standard NaN representation.
    df = df.replace("?", np.nan).copy()
    df = df.dropna(subset=["readmitted"]).reset_index(drop=True)

    # Preserve patient_nbr temporarily so the split can be patient-group aware.
    group_ids = df["patient_nbr"].to_numpy()
    y = df["readmitted"].map({"NO": 0, ">30": 1, "<30": 2}).astype(int).to_numpy()

    # Remove identifiers and the variables explicitly documented as unreliable/highly missing.
    # Remove identifiers and documented high-missingness fields from model input.
    drop_columns = ["encounter_id", "weight", "payer_code"]
    X = df.drop(columns=["readmitted"] + drop_columns, errors="ignore")

    # Use stratified group folds so the same patient cannot cross data partitions.
    sgkf = StratifiedGroupKFold(n_splits=10, shuffle=True, random_state=seed)
    folds = list(sgkf.split(X, y, groups=group_ids))

    # Use disjoint patient groups for test, validation, and training.
    test_idx = folds[0][1]
    val_idx = folds[1][1]
    train_idx = np.concatenate([folds[k][1] for k in range(2, 10)])

    if run_mode == "pilot":
        train_limit = min(PILOT_LIMITS["diabetes_train"], len(train_idx))
        val_limit = min(PILOT_LIMITS["diabetes_val"], len(val_idx))
        # Reduce only the development workload; the test set remains untouched.
        train_idx, _ = train_test_split(
            train_idx,
            train_size=train_limit,
            stratify=y[train_idx],
            random_state=seed,
        )
        val_idx, _ = train_test_split(
            val_idx,
            train_size=val_limit,
            stratify=y[val_idx],
            random_state=seed,
        )

    X_train_df = X.iloc[train_idx].copy()
    X_val_df = X.iloc[val_idx].copy()
    X_test_df = X.iloc[test_idx].copy()

    y_train = y[train_idx]
    y_val = y[val_idx]
    y_test = y[test_idx]

    # Keep categorical codes categorical instead of treating them as continuous measurements.
    numeric_cols = X_train_df.select_dtypes(include=[np.number]).columns.tolist()
    categorical_cols = [c for c in X_train_df.columns if c not in numeric_cols]
    for col in ["admission_type_id", "discharge_disposition_id", "admission_source_id"]:
        if col in numeric_cols:
            numeric_cols.remove(col)
            categorical_cols.append(col)

    num_pipe = Pipeline([
        ("imputer", SimpleImputer(strategy="median")),
        ("scaler", StandardScaler()),
    ])
    cat_pipe = Pipeline([
        ("imputer", SimpleImputer(strategy="most_frequent")),
        ("encoder", OrdinalEncoder(handle_unknown="use_encoded_value", unknown_value=-1)),
        ("scaler", StandardScaler()),
    ])

    preprocessor = ColumnTransformer([
        ("num", num_pipe, numeric_cols),
        ("cat", cat_pipe, categorical_cols),
    ])

    # Fit preprocessing on training data only to prevent data leakage.
    X_train = preprocessor.fit_transform(X_train_df).astype(np.float32)
    X_val = preprocessor.transform(X_val_df).astype(np.float32)
    X_test = preprocessor.transform(X_test_df).astype(np.float32)

    # Conv1D consumes (samples, sequence_length, channels).
    X_train = X_train[..., None]
    X_val = X_val[..., None]
    X_test = X_test[..., None]

    # Compute class weights from training labels only.
    classes = np.unique(y_train)
    weights = compute_class_weight(class_weight="balanced", classes=classes, y=y_train)
    class_weight = {int(c): float(w) for c, w in zip(classes, weights)}

    # Verify that patient identifiers do not overlap across partitions.
    train_patients = set(group_ids[train_idx])
    val_patients = set(group_ids[val_idx])
    test_patients = set(group_ids[test_idx])
    overlaps = {
        "train_val": len(train_patients & val_patients),
        "train_test": len(train_patients & test_patients),
        "val_test": len(val_patients & test_patients),
    }

    return {
        "X_train": X_train,
        "y_train": y_train,
        "X_val": X_val,
        "y_val": y_val,
        "X_test": X_test,
        "y_test": y_test,
        "preprocessor": preprocessor,
        "class_weight": class_weight,
        "dropped_columns": drop_columns,
        "feature_count": X_train.shape[1],
        "numeric_cols": numeric_cols,
        "categorical_cols": categorical_cols,
        "patient_overlap": overlaps,
    }


diabetes_data = prepare_diabetes(diabetes_raw, SEED, RUN_MODE)
X_diab_tr, y_diab_tr = diabetes_data["X_train"], diabetes_data["y_train"]
X_diab_val, y_diab_val = diabetes_data["X_val"], diabetes_data["y_val"]
X_diab_test, y_diab_test = diabetes_data["X_test"], diabetes_data["y_test"]
DIABETES_NAMES = ["NO", ">30", "<30"]

print("Train:", X_diab_tr.shape, y_diab_tr.shape)
print("Val  :", X_diab_val.shape, y_diab_val.shape)
print("Test :", X_diab_test.shape, y_diab_test.shape)
print("Encoded features:", diabetes_data["feature_count"])
print("Dropped columns:", diabetes_data["dropped_columns"])
print("Patient overlaps:", diabetes_data["patient_overlap"])
print("Class weights:", diabetes_data["class_weight"])


````

## 6.2 Feature distributions and target imbalance

The purpose of the exploratory distributions is not to decorate the notebook. They provide evidence for decisions made later in the pipeline.

### Numeric distributions

Hospital-related variables can have very different scales and skewness. A feature representing a count may concentrate near zero, while another variable may occupy a broader range. Neural-network optimization can be unnecessarily difficult when such scales differ strongly.

The standardization step therefore transforms training features using:

\[
z = \frac{x - \mu_{train}}{\sigma_{train}}
\]

where the mean and standard deviation are estimated from the training partition only.

### Target distribution

The three `readmitted` classes are not equally frequent. This imbalance is why the training loss uses training-derived class weights and why macro-level metrics are reported alongside standard accuracy.

The report should never interpret a good overall accuracy as proof that all three clinical outcomes are being recognized equally well. Per-class recall and the confusion matrix provide that missing detail.

### 6.2.1 Why class weights are reported

The training loss can be written conceptually as:

\[
L = - w_y \log p(y|x)
\]

where `w_y` is the training-derived weight for the true class. The effect is to increase the contribution of under-represented classes to the optimization objective.

The test metrics themselves are still reported on the original test distribution. Class weights change how the model is trained; they do not change the underlying evaluation samples.



# Part E - Shared evaluation and visualization methodology

## 7. Evaluation: what counts as evidence?

A machine-learning experiment should answer more than "how accurate is it?". A serious evaluation asks whether the model generalizes, whether its performance is distributed across classes, whether its probability estimates are reasonable, and what types of errors remain.

### Core metrics

**Accuracy** is the proportion of correctly classified samples.

**Precision** asks, for a predicted class, how often the prediction is correct.

**Recall** asks, for a true class, how many of its examples are detected.

**F1-score** is the harmonic mean of precision and recall and is useful when both false positives and false negatives matter.

**Macro F1** averages class-level F1 equally across classes.

**Weighted F1** weights each class by its number of samples.

**Log loss** evaluates the quality of the entire predicted probability distribution. A model can have the same top-1 accuracy as another model while having worse log loss because it is confidently wrong.

**Top-5 accuracy** is included for datasets with at least five classes, especially CIFAR-100.

### Calibration diagnostics

The notebook also compares confidence against empirical accuracy using a reliability diagram and computes expected calibration error (ECE). These diagnostics answer a different question from accuracy: when the model says it is 90% confident, does it behave like a 90%-reliable predictor on average?

### Error analysis

The confusion matrix and misclassified examples provide the qualitative side of the report. Repeated errors can reveal:

- visually similar categories;
- classes with insufficient discriminative information;
- systematic class imbalance effects;
- high-confidence mistakes;
- behavior that differs substantially between CNN3 and CNN5.




````python
# Purpose: Compute consistent metrics and diagnostics for TensorFlow and PyTorch predictions.
def top_k_accuracy(y_true, probs, k=5):
    """Compute top-k classification accuracy.

    Purpose:
        Measure whether the true class appears among the model's k highest-probability classes.

    Parameters:
        y_true (np.ndarray): Integer target labels.
        probs (np.ndarray): Class probabilities with shape (N, C).
        k (int): Number of highest-probability candidates to consider.

    Returns:
        float: Fraction of samples whose true class is inside the top-k set.
    """
    k = min(k, probs.shape[1])
    topk = np.argsort(probs, axis=1)[:, -k:]
    return float(np.mean([y in row for y, row in zip(y_true, topk)]))


def ece_score(y_true, probs, n_bins=10):
    """Compute expected calibration error using confidence bins.

    Purpose:
        Quantify the gap between average predicted confidence and empirical accuracy.

    Parameters:
        y_true (np.ndarray): Integer target labels.
        probs (np.ndarray): Class probabilities.
        n_bins (int): Number of confidence intervals.

    Returns:
        float: Expected calibration error in the range approximately 0 to 1.
    """
    conf = probs.max(axis=1)
    pred = probs.argmax(axis=1)
    correct = pred == y_true
    edges = np.linspace(0, 1, n_bins + 1)
    ece = 0.0
    for i in range(n_bins):
        lo, hi = edges[i], edges[i+1]
        mask = (conf >= lo) & ((conf < hi) if i < n_bins-1 else (conf <= hi))
        if mask.any():
            # Weight each bin by its frequency so large bins contribute proportionally.
            ece += mask.mean() * abs(correct[mask].mean() - conf[mask].mean())
    return float(ece)


def evaluate_predictions(y_true, probs, class_names):
    """Compute the common multiclass metric set used by the final comparison table.

    Purpose:
        Ensure both frameworks are evaluated with identical metric definitions.

    Parameters:
        y_true (np.ndarray): Ground-truth integer labels.
        probs (np.ndarray): Predicted class probabilities.
        class_names (list): Names used to determine the number of classes for log loss.

    Returns:
        tuple: Dictionary of scalar metrics and the argmax class predictions.
    """
    # Convert probability vectors into the final top-1 class prediction.
    preds = probs.argmax(axis=1)
    p_m, r_m, f1_m, _ = precision_recall_fscore_support(y_true, preds, average="macro", zero_division=0)
    p_w, r_w, f1_w, _ = precision_recall_fscore_support(y_true, preds, average="weighted", zero_division=0)
    out = {
        "accuracy": float(accuracy_score(y_true, preds)),
        "precision_macro": float(p_m),
        "recall_macro": float(r_m),
        "f1_macro": float(f1_m),
        "precision_weighted": float(p_w),
        "recall_weighted": float(r_w),
        "f1_weighted": float(f1_w),
        "log_loss": float(log_loss(y_true, probs, labels=np.arange(len(class_names)))),
        "mean_confidence": float(probs.max(axis=1).mean()),
        "ece": ece_score(y_true, probs),
    }
    if probs.shape[1] >= 5:
        # Top-5 is informative for large class spaces such as CIFAR-100.
        out["top5_accuracy"] = top_k_accuracy(y_true, probs, 5)
    return out, preds


def plot_history(history, title):
    """Plot training and validation loss/accuracy across epochs.

    Purpose:
        Diagnose convergence, plateaus, and potential overfitting from the training trajectory.

    Parameters:
        history (dict): Training history containing loss/accuracy and validation counterparts.
        title (str): Figure title prefix.

    Returns:
        None. Two diagnostic plots are displayed.
    """
    epochs = np.arange(1, len(history["loss"])+1)
    fig, ax = plt.subplots(1, 2, figsize=(12, 4))
    ax[0].plot(epochs, history["loss"], marker="o", label="train")
    ax[0].plot(epochs, history["val_loss"], marker="o", label="val")
    ax[0].set_title(title + " - loss"); ax[0].set_xlabel("Epoch"); ax[0].set_ylabel("Loss"); ax[0].legend()
    ax[1].plot(epochs, history["accuracy"], marker="o", label="train")
    ax[1].plot(epochs, history["val_accuracy"], marker="o", label="val")
    ax[1].set_title(title + " - accuracy"); ax[1].set_xlabel("Epoch"); ax[1].set_ylabel("Accuracy"); ax[1].legend()
    plt.tight_layout(); plt.show()


def plot_confusion(y_true, preds, class_names, title, dataset_key):
    """Plot an annotated confusion matrix and a readable subset for large class spaces.

    Purpose:
        Show exact error counts and, for 100-class tasks, highlight the strongest confusion region.

    Parameters:
        y_true (np.ndarray): Ground-truth labels.
        preds (np.ndarray): Predicted labels.
        class_names (list): Human-readable class names.
        title (str): Model/dataset label used in the figure title.
        dataset_key (str): File-name-safe dataset identifier.

    Returns:
        np.ndarray: Full confusion matrix counts.
    """
    cm = confusion_matrix(y_true, preds)
    n = len(class_names)
    fig, ax = plt.subplots(figsize=(8 if n <= 20 else 12, 8 if n <= 20 else 12))
    im = ax.imshow(cm, cmap="Blues")
    plt.colorbar(im, ax=ax)
    ax.set_xlabel("Predicted"); ax.set_ylabel("True"); ax.set_title(title + " - count confusion matrix")
    ax.set_xticks(range(n)); ax.set_yticks(range(n))
    if n <= 20:
        ax.set_xticklabels(class_names, rotation=45, ha="right", fontsize=8)
        ax.set_yticklabels(class_names, fontsize=8)
        threshold = cm.max() * 0.5 if cm.size else 0
        for i in range(n):
            for j in range(n):
                # Annotate every cell for small class spaces so exact counts are visible.
                ax.text(j, i, str(cm[i,j]), ha="center", va="center", fontsize=7,
                        color="white" if cm[i,j] > threshold else "black")
    else:
        ax.set_xticklabels(range(n), fontsize=5); ax.set_yticklabels(range(n), fontsize=5)
    fig.savefig(FIGURES_DIR / f"{dataset_key}_{title.replace(' ', '_').lower()}_cm.png", dpi=160, bbox_inches="tight")
    plt.show()

    if n > 20:
        # Select the classes with the largest off-diagonal confusion mass for readability.
        off = cm.copy(); np.fill_diagonal(off, 0)
        important = np.argsort(off.sum(axis=1))[-20:]
        sub = cm[np.ix_(important, important)]
        names = [class_names[i] for i in important]
        fig, ax = plt.subplots(figsize=(12, 10))
        im = ax.imshow(sub, cmap="Blues"); plt.colorbar(im, ax=ax)
        ax.set_xticks(range(20)); ax.set_yticks(range(20))
        ax.set_xticklabels(names, rotation=60, ha="right", fontsize=7); ax.set_yticklabels(names, fontsize=7)
        for i in range(20):
            for j in range(20):
                # Keep numeric counts visible for the focused confusion subset.
                ax.text(j, i, str(sub[i,j]), ha="center", va="center", fontsize=6,
                        color="white" if sub[i,j] > sub.max()*0.5 else "black")
        ax.set_title(title + " - annotated top confusion subset")
        plt.tight_layout(); plt.show()
    return cm


def plot_per_class_metrics(y_true, preds, class_names, title):
    """Plot precision, recall, and F1 for every class.

    Purpose:
        Detect classes that are hidden by an otherwise strong aggregate metric.

    Parameters:
        y_true (np.ndarray): Ground-truth labels.
        preds (np.ndarray): Predicted labels.
        class_names (list): Human-readable labels.
        title (str): Figure title prefix.

    Returns:
        pd.DataFrame: Per-class precision/recall/F1 table.
    """
    report = classification_report(y_true, preds, target_names=class_names, output_dict=True, zero_division=0)
    df = pd.DataFrame(report).T.iloc[:len(class_names)]
    ax = df[["precision", "recall", "f1-score"]].plot(kind="bar", figsize=(12, max(5, len(class_names)*0.18)))
    ax.set_ylim(0, 1.05); ax.set_title(title + " - per-class metrics"); ax.set_ylabel("Score")
    if len(class_names) <= 20:
        ax.tick_params(axis="x", rotation=35)
    else:
        ax.set_xticks([])
    plt.tight_layout(); plt.show()
    return df


def plot_confidence(y_true, probs, title):
    """Plot confidence distributions for correct and incorrect predictions.

    Purpose:
        Identify whether the model tends to be uncertain or highly confident when making mistakes.

    Parameters:
        y_true (np.ndarray): Ground-truth labels.
        probs (np.ndarray): Predicted probabilities.
        title (str): Figure title prefix.

    Returns:
        None. Histogram is displayed.
    """
    pred = probs.argmax(axis=1)
    conf = probs.max(axis=1)
    correct = pred == y_true
    plt.figure(figsize=(9,4))
    plt.hist(conf[correct], bins=20, alpha=0.7, label="Correct")
    plt.hist(conf[~correct], bins=20, alpha=0.7, label="Incorrect")
    plt.xlabel("Confidence"); plt.ylabel("Count"); plt.title(title + " - confidence distribution"); plt.legend(); plt.tight_layout(); plt.show()


def plot_reliability(y_true, probs, title, n_bins=10):
    """Plot mean confidence against empirical accuracy across confidence bins.

    Purpose:
        Visualize probability calibration and compare model confidence with actual reliability.

    Parameters:
        y_true (np.ndarray): Ground-truth labels.
        probs (np.ndarray): Predicted probabilities.
        title (str): Figure title prefix.
        n_bins (int): Number of confidence bins.

    Returns:
        None. Reliability diagram is displayed.
    """
    conf = probs.max(axis=1); pred = probs.argmax(axis=1); correct = pred == y_true
    edges = np.linspace(0, 1, n_bins+1); bin_c=[]; bin_a=[]
    for i in range(n_bins):
        mask = (conf >= edges[i]) & ((conf < edges[i+1]) if i < n_bins-1 else (conf <= edges[i+1]))
        if mask.any():
            bin_c.append(conf[mask].mean()); bin_a.append(correct[mask].mean())
    plt.figure(figsize=(6,6)); plt.plot([0,1],[0,1], "--", label="Perfect calibration"); plt.plot(bin_c, bin_a, "o-", label="Model")
    plt.xlabel("Mean confidence"); plt.ylabel("Accuracy"); plt.title(title + " - reliability diagram"); plt.legend(); plt.tight_layout(); plt.show()


def plot_misclassified(images, y_true, preds, probs, class_names, title, n=12):
    """Display a small set of wrong predictions with true/predicted labels and confidence.

    Purpose:
        Provide concrete examples for qualitative error analysis.

    Parameters:
        images (np.ndarray): Test images aligned with y_true.
        y_true (np.ndarray): True labels.
        preds (np.ndarray): Predicted labels.
        probs (np.ndarray): Predicted probabilities.
        class_names (list): Readable class names.
        title (str): Figure title prefix.
        n (int): Maximum number of mistakes to show.

    Returns:
        None. Error examples are displayed when available.
    """
    wrong = np.where(y_true != preds)[0][:n]
    if len(wrong) == 0:
        print("No misclassified examples."); return
    cols = 4; rows = math.ceil(len(wrong)/cols)
    plt.figure(figsize=(12,3*rows))
    for k, idx in enumerate(wrong):
        ax = plt.subplot(rows, cols, k+1)
        ax.imshow(images[idx], cmap="gray" if images[idx].ndim == 2 else None)
        ax.set_title(f"True: {class_names[y_true[idx]]}\nPred: {class_names[preds[idx]]} | {probs[idx].max():.3f}")
        ax.axis("off")
    plt.suptitle(title + " - error examples"); plt.tight_layout(); plt.show()


def top_confusion_pairs(y_true, preds, class_names, top_n=15):
    """Return the strongest off-diagonal confusion pairs.

    Purpose:
        Turn a large confusion matrix into a readable list of recurring class-to-class errors.

    Parameters:
        y_true (np.ndarray): Ground-truth labels.
        preds (np.ndarray): Predicted labels.
        class_names (list): Readable class names.
        top_n (int): Number of pairs to return.

    Returns:
        pd.DataFrame: Columns are count, true_class, and predicted_class.
    """
    cm = confusion_matrix(y_true, preds)
    pairs=[]
    for i in range(len(class_names)):
        for j in range(len(class_names)):
            if i != j and cm[i,j] > 0:
                pairs.append((int(cm[i,j]), class_names[i], class_names[j]))
    pairs.sort(reverse=True)
    return pd.DataFrame(pairs[:top_n], columns=["count","true_class","predicted_class"])


def save_experiment(dataset_key, framework, depth, y_true, probs, metrics):
    """Persist predictions, probabilities, metrics, and confusion counts for one run.

    Purpose:
        Make every experiment auditable and ready for later web-demo integration.

    Parameters:
        dataset_key (str): File-safe dataset identifier.
        framework (str): `tensorflow` or `pytorch`.
        depth (int): CNN convolutional depth (3 or 5).
        y_true (np.ndarray): Ground-truth test labels.
        probs (np.ndarray): Test class probabilities.
        metrics (dict): Scalar metric dictionary.

    Returns:
        None. Files are written under the project output directories.
    """
    prefix = f"{dataset_key}_{framework}_cnn{depth}"
    # Save the full probability vector, not only the final class, so confidence can be analyzed later.
    np.savez_compressed(PREDICTIONS_DIR / f"{prefix}_predictions.npz", y_true=y_true, y_pred=probs.argmax(1), probabilities=probs)
    with open(METRICS_DIR / f"{prefix}.json", "w", encoding="utf-8") as f:
        json.dump(metrics, f, indent=2)
    pd.DataFrame(confusion_matrix(y_true, probs.argmax(1))).to_csv(METRICS_DIR / f"{prefix}_confusion_matrix.csv", index=False)


def evaluate_and_visualize(dataset_key, framework, depth, y_true, probs, class_names, images=None, history=None):
    """Run the complete standardized evaluation and visualization workflow.

    Purpose:
        Keep the evaluation pipeline identical across TensorFlow and PyTorch experiments.

    Parameters:
        dataset_key (str): Dataset identifier.
        framework (str): Framework name.
        depth (int): CNN convolutional depth.
        y_true (np.ndarray): Ground-truth labels.
        probs (np.ndarray): Test probabilities.
        class_names (list): Readable class names.
        images (np.ndarray|None): Test images for visual error examples.
        history (dict|None): Training history for convergence plots.

    Returns:
        tuple: Scalar metrics dictionary and predicted class labels.
    """
    metrics, preds = evaluate_predictions(y_true, probs, class_names)
    metrics.update({"dataset": dataset_key, "framework": framework, "conv_layers": depth})
    print(f"\n{dataset_key} | {framework} | CNN{depth}")
    display(pd.DataFrame([metrics]).T.rename(columns={0:"value"}))
    if history is not None:
        # Training curves diagnose optimization behavior before test metrics are interpreted.
        plot_history(history, f"{dataset_key} {framework} CNN{depth}")
    plot_confusion(y_true, preds, class_names, f"{dataset_key} {framework} CNN{depth}", dataset_key)
    print("Per-class metrics:")
    display(plot_per_class_metrics(y_true, preds, class_names, f"{dataset_key} {framework} CNN{depth}"))
    plot_confidence(y_true, probs, f"{dataset_key} {framework} CNN{depth}")
    plot_reliability(y_true, probs, f"{dataset_key} {framework} CNN{depth}")
    if images is not None:
        plot_misclassified(images, y_true, preds, probs, class_names, f"{dataset_key} {framework} CNN{depth}")
    print("Top confusion pairs:")
    display(top_confusion_pairs(y_true, preds, class_names))
    save_experiment(dataset_key, framework, depth, y_true, probs, metrics)
    return metrics, preds


````

### 7.1 Calibration and confidence are different from accuracy

Accuracy answers whether the top-1 class is correct. Confidence answers how strongly the model prefers its top class. A model can therefore have high accuracy but poor calibration, or modest accuracy but relatively honest confidence.

The reliability diagram compares these two quantities across bins. ECE compresses the same idea into one summary number. Neither should replace the main classification metrics; they complement them.



## 7.0 Experimental hyperparameters and model-design rationale

Hyperparameters determine how the optimization problem is solved. They should therefore be explained rather than treated as arbitrary numbers.

| Hyperparameter | Value | Purpose and interpretation |
|---|---:|---|
| Convolution kernel | 3 x 3 for images / 3 for Conv1D | Captures local neighborhoods while keeping parameter count manageable |
| Initial learning rate | 0.001 | Controls the size of optimizer updates at the beginning of training |
| Optimizer | Adam | Adaptive gradient-based optimization used consistently across frameworks |
| Batch size | 128 | Balances gradient stability, memory usage, and update frequency |
| Maximum epochs | 12 | Upper bound; early stopping can terminate earlier |
| Dropout | 0.30 | Regularization that randomly removes classifier inputs during training |
| Batch Normalization | Enabled | Stabilizes intermediate activations and often makes optimization easier |
| LR reduction factor | 0.5 | Halves the learning rate when validation progress stalls |
| Minimum learning rate | 1e-6 | Prevents the scheduler from reducing the learning rate toward zero |
| Early-stopping patience | 3 | Allows temporary validation fluctuations before stopping |
| Checkpoint criterion | Lowest validation loss | Selects the model according to generalization on held-out validation data |

### Why the same settings are reused

A5 compares depth and frameworks. If every experiment receives a separate manually tuned learning rate, the study becomes difficult to interpret because the observed difference may come from hyperparameter tuning rather than from depth or framework structure.

The notebook therefore uses the same initial optimization policy and the same conceptual architecture schedule for the corresponding TensorFlow and PyTorch models. The final report should acknowledge that framework implementations can still exhibit small runtime or numerical differences.




````python
# Purpose: Calculate output sizes before training so the classifier dimensions are auditable.
def convolution_output_size(size, kernel=3, stride=1, padding=1):
    """Compute the output length of one convolution dimension.

    Purpose:
        Apply the standard convolution output-size formula before constructing the model.

    Parameters:
        size (int): Input spatial or sequence size.
        kernel (int): Kernel size.
        stride (int): Convolution stride.
        padding (int): Padding size on each side.

    Returns:
        int: Output size after convolution.
    """
    return (size + 2 * padding - kernel) // stride + 1


def print_shape_plan(input_size, pool_after_layers, depth):
    """Print the spatial/sequence size after every convolutional block.

    Purpose:
        Make tensor-shape reasoning explicit before the model is trained.

    Parameters:
        input_size (int): Initial height, width, or sequence length.
        pool_after_layers (set): Zero-based convolution indices followed by pooling.
        depth (int): Number of convolutional layers.

    Returns:
        None. Shape transitions are printed to the notebook.
    """
    size = input_size
    print(f"Input size: {size}")
    for layer_id in range(depth):
        # Convolution with padding=1 and kernel=3 preserves the current size.
        size = convolution_output_size(size)
        if layer_id in pool_after_layers:
            # Pool size 2 with stride 2 halves the spatial/sequence dimension approximately.
            size //= 2
        print(f"After Conv{layer_id + 1}" + (" + Pool" if layer_id in pool_after_layers else "") + f": {size}")

print("Fashion-MNIST shape plan for CNN3")
print_shape_plan(28, {0, 1}, 3)
print("\nFashion-MNIST shape plan for CNN5")
print_shape_plan(28, {0, 1, 3}, 5)
print("\nCIFAR-100 shape plan for CNN5")
print_shape_plan(32, {0, 1, 3}, 5)


````

# Part F - TensorFlow / Keras implementation

## 8. TensorFlow CNN architecture and abstraction level

TensorFlow/Keras represents the CNN as a sequence of reusable layers. This is exactly the abstraction change emphasized in Lecture 04.

Conceptually:

```text
Convolution → ReLU → Pooling → ... → Global aggregation → Classifier
```

becomes a Keras model built from `Conv2D`, `BatchNormalization`, `ReLU`, `MaxPooling2D`, `GlobalAveragePooling2D`, `Dropout`, and `Dense` layers. For the tabular dataset, the same idea is expressed with the 1-D counterparts.

### Why Batch Normalization is used

Batch Normalization standardizes intermediate activations using batch statistics during training and running estimates during inference. In this experiment it is used as an optimization aid after convolution and before ReLU.

### Why Global Average Pooling is used

Instead of flattening a large feature map into a very large dense input, global average pooling reduces each feature map to one average value. This dramatically reduces the number of classifier parameters and makes the comparison of CNN3 and CNN5 focus more on convolutional representation capacity.

### Why dropout is placed before the classifier

Dropout is used as a regularizer. During training, a random subset of classifier inputs is suppressed, reducing reliance on a small set of activation patterns. It is disabled automatically during evaluation.




````python
# Purpose: Define, train, and use TensorFlow/Keras CNN3 and CNN5 models for all A5 datasets.
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers


def build_tf_cnn(input_shape, num_classes, depth=3, task_type="image"):
    """Build the A5 CNN family in TensorFlow/Keras.

    Purpose:
        Construct a controlled CNN3 or CNN5 architecture for either image data (Conv2D)
        or the assignment-driven tabular adaptation (Conv1D).

    Parameters:
        input_shape (tuple): Shape excluding the batch dimension.
        num_classes (int): Number of output classes.
        depth (int): Number of convolutional layers; only 3 or 5 is allowed.
        task_type (str): `image` for Conv2D or `tabular` for Conv1D.

    Returns:
        keras.Model: Compiled-later Keras model architecture.
    """
    if depth not in (3,5):
        raise ValueError("depth must be 3 or 5")

    # Keras Input makes the expected tensor shape explicit before any layer is added.
    inputs = keras.Input(shape=input_shape)
    x = inputs
    filters = [32,64,128,128,160] if depth == 5 else [32,64,128]
    pool_after = {0,1} if depth == 3 else {0,1,3}

    if task_type == "image":
        # Use the same light augmentation policy for every image model: random horizontal flip.
        x = layers.RandomFlip("horizontal")(x)
        for i in range(depth):
            # BatchNorm stabilizes intermediate activations; ReLU supplies non-linearity.
            x = layers.Conv2D(filters[i], 3, padding="same", use_bias=False)(x)
            x = layers.BatchNormalization()(x)
            x = layers.ReLU()(x)
            if i in pool_after:
                # Pooling reduces spatial resolution and the amount of later computation.
                x = layers.MaxPooling2D(2)(x)
        # Global pooling avoids a large dense layer tied to the spatial dimensions.
        x = layers.GlobalAveragePooling2D()(x)
    elif task_type == "tabular":
        for i in range(depth):
            # Conv1D scans local windows of the processed feature sequence.
            x = layers.Conv1D(filters[i], 3, padding="same", use_bias=False)(x)
            x = layers.BatchNormalization()(x)
            x = layers.ReLU()(x)
            if i in pool_after:
                x = layers.MaxPooling1D(2)(x)
        # Global pooling converts the final sequence representation into one vector per sample.
        x = layers.GlobalAveragePooling1D()(x)
    else:
        raise ValueError("task_type must be image or tabular")

    # Dropout regularizes the classifier input during training.
    x = layers.Dropout(0.30)(x)
    # The output layer returns logits; softmax is applied later during evaluation.
    outputs = layers.Dense(num_classes)(x)
    return keras.Model(inputs, outputs)


def train_tf(model, X_train, y_train, X_val, y_val, class_weight, model_path, log_dir):
    """Train a Keras model with validation monitoring and best-checkpoint recovery.

    Purpose:
        Encapsulate compilation, optimization callbacks, timing, and checkpoint selection
        while keeping the training configuration consistent across all TensorFlow experiments.

    Parameters:
        model (keras.Model): CNN model returned by `build_tf_cnn`.
        X_train, y_train: Training tensors and labels.
        X_val, y_val: Validation tensors and labels.
        class_weight (dict|None): Training-derived class weights, if needed.
        model_path (Path): Destination for the best `.keras` checkpoint.
        log_dir (Path): TensorBoard directory when TensorBoard is available.

    Returns:
        dict: Training/validation history plus measured training time.
    """
    # Compile uses the same learning rate, loss family, and optimizer policy for all models.
    model.compile(
        optimizer=keras.optimizers.Adam(learning_rate=LEARNING_RATE),
        loss=keras.losses.SparseCategoricalCrossentropy(from_logits=True),
        metrics=[keras.metrics.SparseCategoricalAccuracy(name="accuracy")],
    )

    callbacks = [
        # Stop after repeated validation stagnation and restore the best validation weights.
        keras.callbacks.EarlyStopping(monitor="val_loss", patience=3, restore_best_weights=True),
        # Reduce the learning rate when progress stalls instead of continuing with large updates.
        keras.callbacks.ReduceLROnPlateau(monitor="val_loss", factor=0.5, patience=2, min_lr=1e-6, verbose=1),
        # Persist the best model so inference uses the selected checkpoint rather than the last epoch.
        keras.callbacks.ModelCheckpoint(model_path, monitor="val_loss", save_best_only=True),
    ]

    # TensorBoard is a monitoring tool, not a training requirement, so it is optional.
    try:
        import tensorboard  # noqa: F401
        callbacks.append(keras.callbacks.TensorBoard(log_dir=log_dir))
        print("TensorBoard logging enabled.")
    except ImportError:
        print("TensorBoard is not installed; continuing without TensorBoard logging.")

    start = time.perf_counter()
    hist = model.fit(
        X_train, y_train,
        validation_data=(X_val, y_val),
        epochs=EPOCHS,
        batch_size=BATCH_SIZE,
        shuffle=True,
        class_weight=class_weight,
        callbacks=callbacks,
        verbose=2,
    )
    out = {k:[float(v) for v in vals] for k, vals in hist.history.items()}
    out["training_seconds"] = float(time.perf_counter() - start)
    out["best_epoch"] = int(np.argmin(out["val_loss"]) + 1)
    return out


def predict_tf(model, X):
    """Run TensorFlow inference and convert logits to NumPy probabilities.

    Purpose:
        Standardize the output format so TensorFlow predictions can use the same
        framework-independent evaluation functions as PyTorch predictions.

    Parameters:
        model (keras.Model): Trained Keras model.
        X (np.ndarray): Test input tensor.

    Returns:
        tuple: Logits array and softmax probability array.
    """
    # Predict in batches to avoid unnecessary peak memory usage on larger datasets.
    logits = model.predict(X, batch_size=TEST_BATCH_SIZE, verbose=0)
    probs = tf.nn.softmax(logits, axis=1).numpy()
    return logits, probs


````

## 8.0 Tensor-shape planning before training

Lecture 04 explicitly requires tensor shapes to be calculated before the final classifier is written. This avoids one of the most common CNN implementation mistakes: giving a `Dense` or `Linear` layer an incorrect input dimension.

For example, for a 28 x 28 Fashion-MNIST image, a 3-layer model that performs pooling after the first two convolutional blocks follows the spatial transformation:

```text
Input             : 28 x 28
Conv 3, padding 1 : 28 x 28
Pool 2            : 14 x 14
Conv 3, padding 1 : 14 x 14
Pool 2            : 7 x 7
Conv 3, padding 1 : 7 x 7
GlobalAvgPool     : 1 x 1 per feature map
```

Because global average pooling is used, the classifier input is the number of final filters rather than `channels x height x width`. This makes the architecture less sensitive to the exact spatial size and keeps the parameter count moderate.

For CIFAR-100 the same principle starts from 32 x 32 x 3, and for Diabetes the Conv1D input starts from `sequence_length x 1`. The notebook prints shape plans before training so that architecture assumptions are visible.

# Part G - PyTorch implementation

## 9. PyTorch CNN architecture and abstraction level

PyTorch uses `nn.Module` to express the same conceptual CNN architecture. This corresponds directly to the second framework implementation described in Lecture 04.

For an image model:

```text
nn.Conv2d
→ nn.BatchNorm2d
→ nn.ReLU
→ nn.MaxPool2d
→ ...
→ nn.AdaptiveAvgPool2d
→ nn.Dropout
→ nn.Linear
```

For the Diabetes experiment, the same pattern uses the 1-D equivalents.

### Tensor layout difference

TensorFlow/Keras normally receives image tensors as:

```text
(batch, height, width, channels)
```

PyTorch convolution layers expect:

```text
(batch, channels, height, width)
```

This is a representation difference, not a different learning problem. The notebook converts the NumPy arrays at the Dataset boundary so that the model itself receives the convention expected by PyTorch.




````python
# Purpose: Define the PyTorch datasets, models, validation loop, inference path, and training loop.
import torch
from torch import nn
from torch.utils.data import Dataset, DataLoader

DEVICE = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print("Device:", DEVICE)

try:
    from torchvision import transforms
    HAS_TORCHVISION = True
except Exception as exc:
    HAS_TORCHVISION = False
    print("torchvision warning:", exc)


class NumpyImageDataset(Dataset):
    """Wrap a NumPy image array in the Dataset interface expected by PyTorch.

    Purpose:
        Convert NumPy HxW/HxWxC images into PyTorch channel-first tensors and optionally
        apply training-only augmentation.
    """
    def __init__(self, X, y, train=False):
        """Initialize the image dataset wrapper and optional training augmentation.

        Purpose:
            Store image/label arrays and configure augmentation without modifying validation/test data.
        """
        # Store the arrays without changing the underlying labels.
        self.X = X
        self.y = y.astype(np.int64)
        self.augment = None
        if HAS_TORCHVISION and train:
            # Match the TensorFlow image policy with a light horizontal flip only.
            self.augment = transforms.Compose([
                transforms.RandomHorizontalFlip(0.5),
            ])

    def __len__(self):
        """Return the number of image examples in this dataset."""
        return len(self.y)

    def __getitem__(self, idx):
        """Return one image tensor and its integer target label.

        Purpose:
            Perform the NumPy-to-PyTorch channel conversion at the data boundary.
        """
        img = self.X[idx]
        if img.ndim == 2:
            # Add the single grayscale channel before transposing to CxHxW.
            img = img[...,None]
        tensor = torch.from_numpy(img.transpose(2,0,1)).float()
        if self.augment is not None:
            # Augmentation is applied only to the training dataset.
            tensor = self.augment(tensor)
        return tensor, torch.tensor(self.y[idx], dtype=torch.long)


class NumpyTabularDataset(Dataset):
    """Wrap the processed Diabetes Conv1D tensor and labels as a PyTorch Dataset."""
    def __init__(self, X, y):
        """Initialize a tabular dataset wrapper around processed Conv1D tensors.

        Purpose:
            Convert processed NumPy arrays into PyTorch tensors once for efficient batching.
        """
        # Convert the complete arrays once so __getitem__ remains lightweight.
        self.X = torch.from_numpy(X).float()
        self.y = torch.from_numpy(y.astype(np.int64)).long()

    def __len__(self):
        """Return the number of tabular training/evaluation examples."""
        return len(self.y)

    def __getitem__(self, idx):
        """Return one Conv1D-ready feature tensor and its class label."""
        return self.X[idx], self.y[idx]


class ImageCNN(nn.Module):
    """Build the A5 CNN3/CNN5 architecture for 2-D image datasets."""
    def __init__(self, in_channels, num_classes, depth=3):
        """Initialize the image CNN modules for the requested convolutional depth.

        Purpose:
            Build the shared CNN3/CNN5 structure used in the PyTorch image experiments.
        """
        super().__init__()
        # Keep the filter schedule identical to TensorFlow so depth is the main architectural variable.
        filters = [32,64,128,128,160] if depth == 5 else [32,64,128]
        pool_after = {0,1} if depth == 3 else {0,1,3}
        blocks=[]
        for i in range(depth):
            blocks += [
                nn.Conv2d(in_channels if i==0 else filters[i-1], filters[i], 3, padding=1, bias=False),
                nn.BatchNorm2d(filters[i]),
                nn.ReLU(inplace=True),
            ]
            if i in pool_after:
                # Pooling reduces spatial resolution and computation after selected blocks.
                blocks.append(nn.MaxPool2d(2))
        self.features = nn.Sequential(*blocks)
        # Adaptive pooling makes the classifier input equal to the final number of filters.
        self.pool = nn.AdaptiveAvgPool2d(1)
        self.dropout = nn.Dropout(0.30)
        self.classifier = nn.Linear(filters[-1], num_classes)

    def forward(self, x):
        """Define the forward data flow from image tensor to class logits."""
        # Feature extractor learns local-to-global visual representations.
        x = self.features(x)
        # Global average pooling removes dependence on the exact remaining spatial dimensions.
        x = self.pool(x).flatten(1)
        # Dropout regularizes the classifier during training.
        return self.classifier(self.dropout(x))


class TabularCNN(nn.Module):
    """Build the assignment-driven Conv1D CNN3/CNN5 architecture for Diabetes."""
    def __init__(self, num_classes, depth=3):
        """Initialize the assignment-driven Conv1D CNN for the Diabetes experiment.

        Purpose:
            Build CNN3/CNN5 using one input channel over the processed feature sequence.
        """
        super().__init__()
        # The first Conv1D receives one channel and scans the feature sequence.
        filters = [32,64,128,128,160] if depth == 5 else [32,64,128]
        pool_after = {0,1} if depth == 3 else {0,1,3}
        blocks=[]
        for i in range(depth):
            blocks += [
                nn.Conv1d(1 if i==0 else filters[i-1], filters[i], 3, padding=1, bias=False),
                nn.BatchNorm1d(filters[i]),
                nn.ReLU(inplace=True),
            ]
            if i in pool_after:
                blocks.append(nn.MaxPool1d(2))
        self.features = nn.Sequential(*blocks)
        self.pool = nn.AdaptiveAvgPool1d(1)
        self.dropout = nn.Dropout(0.30)
        self.classifier = nn.Linear(filters[-1], num_classes)

    def forward(self, x):
        """Define the forward data flow from feature sequence to class logits."""
        # Apply the Conv1D feature extractor over the processed clinical feature sequence.
        x = self.features(x)
        x = self.pool(x).flatten(1)
        return self.classifier(self.dropout(x))


def loader(ds, shuffle=False):
    """Create a reproducible DataLoader for training, validation, or testing.

    Purpose:
        Provide consistent batching and optional shuffling across every PyTorch experiment.

    Parameters:
        ds (Dataset): Dataset instance to batch.
        shuffle (bool): Whether training samples should be shuffled each epoch.

    Returns:
        DataLoader: Batched PyTorch data iterator.
    """
    # Pin memory only when CUDA is available because it can speed host-to-device transfers.
    return DataLoader(ds, batch_size=BATCH_SIZE, shuffle=shuffle, num_workers=0, pin_memory=torch.cuda.is_available())


def val_loss_acc(model, dl, criterion):
    """Evaluate validation loss and accuracy without updating model parameters.

    Purpose:
        Provide the validation signal used by the scheduler and checkpoint-selection logic.

    Parameters:
        model (nn.Module): Current PyTorch model.
        dl (DataLoader): Validation data loader.
        criterion: Loss function.

    Returns:
        tuple: Mean validation loss and validation accuracy.
    """
    model.eval()
    total_loss=0.0; total=0; correct=0
    with torch.no_grad():
        for xb,yb in dl:
            xb,yb=xb.to(DEVICE),yb.to(DEVICE)
            logits=model(xb); loss=criterion(logits,yb)
            # Accumulate sample-weighted loss so batches of different sizes are handled correctly.
            total_loss += loss.item()*yb.size(0)
            total += yb.size(0)
            correct += (logits.argmax(1)==yb).sum().item()
    return total_loss/max(total,1), correct/max(total,1)


def predict_torch(model, dl):
    """Run PyTorch inference and return labels, logits, and probabilities as NumPy arrays.

    Purpose:
        Convert framework-specific model output into the common format used by the evaluation code.

    Parameters:
        model (nn.Module): Trained model.
        dl (DataLoader): Test data loader.

    Returns:
        tuple: y_true, logits, and softmax probabilities as NumPy arrays.
    """
    model.eval(); logits_all=[]; y_all=[]
    with torch.no_grad():
        for xb,yb in dl:
            # Disable gradients because test inference does not update model parameters.
            logits_all.append(model(xb.to(DEVICE)).cpu().numpy())
            y_all.append(yb.numpy())
    logits=np.concatenate(logits_all); y_true=np.concatenate(y_all)
    # Apply numerically stable softmax after moving logits back to NumPy.
    p=np.exp(logits-logits.max(axis=1,keepdims=True)); p/=p.sum(axis=1,keepdims=True)
    return y_true, logits, p


def train_torch(model, train_dl, val_dl, class_weight, path):
    """Train a PyTorch CNN with Adam, scheduling, gradient clipping, and checkpointing.

    Purpose:
        Make the learning loop explicit, matching the training-cycle explanation in Lecture 04.

    Parameters:
        model (nn.Module): CNN3 or CNN5 model.
        train_dl (DataLoader): Training data.
        val_dl (DataLoader): Validation data.
        class_weight (dict): Training-derived class weights.
        path (Path): Destination for the best `.pth` checkpoint.

    Returns:
        dict: Training/validation history, elapsed time, and best epoch.
    """
    # Convert training-derived class weights into a device-resident tensor for CrossEntropyLoss.
    weights=torch.tensor([class_weight.get(i,1.0) for i in range(len(class_weight))], dtype=torch.float32, device=DEVICE)
    criterion=nn.CrossEntropyLoss(weight=weights)
    optimizer=torch.optim.Adam(model.parameters(), lr=LEARNING_RATE)
    scheduler=torch.optim.lr_scheduler.ReduceLROnPlateau(optimizer, mode="min", factor=0.5, patience=2, min_lr=1e-6)
    history={"loss":[],"val_loss":[],"accuracy":[],"val_accuracy":[]}
    best=float("inf"); bad=0; best_epoch=1; start=time.perf_counter()

    for epoch in range(1,EPOCHS+1):
        model.train(); running=0.0; total=0; correct=0
        for xb,yb in train_dl:
            xb,yb=xb.to(DEVICE),yb.to(DEVICE)
            # Clear old gradients before calculating the next batch gradient.
            optimizer.zero_grad(set_to_none=True)
            # Forward pass produces class logits.
            logits=model(xb)
            # Compute the supervised training objective.
            loss=criterion(logits,yb)
            # Autograd calculates gradients for every learnable parameter.
            loss.backward()
            # Clip very large gradients to improve training stability.
            torch.nn.utils.clip_grad_norm_(model.parameters(),5.0)
            # Apply the optimizer update using the new gradients.
            optimizer.step()
            running += loss.item()*yb.size(0); total += yb.size(0); correct += (logits.argmax(1)==yb).sum().item()

        tr_loss=running/max(total,1); tr_acc=correct/max(total,1)
        vl_loss,vl_acc=val_loss_acc(model,val_dl,criterion)
        # Scheduler observes validation loss rather than training loss because generalization is the target.
        scheduler.step(vl_loss)
        for k,v in [("loss",tr_loss),("accuracy",tr_acc),("val_loss",vl_loss),("val_accuracy",vl_acc)]: history[k].append(float(v))
        print(f"Epoch {epoch:02d} | train loss {tr_loss:.4f} acc {tr_acc:.4f} | val loss {vl_loss:.4f} acc {vl_acc:.4f}")

        if vl_loss < best-1e-5:
            # Save the best validation checkpoint rather than blindly keeping the last epoch.
            best=vl_loss; bad=0; best_epoch=epoch
            torch.save({"model_state_dict":model.state_dict(),"best_val_loss":best,"best_epoch":best_epoch}, path)
        else:
            bad += 1
            if bad >= 3:
                print("Early stopping.")
                break

    # Restore the best checkpoint before any test-set evaluation.
    ckpt=torch.load(path, map_location=DEVICE)
    model.load_state_dict(ckpt["model_state_dict"])
    history["training_seconds"]=float(time.perf_counter()-start)
    history["best_epoch"]=int(ckpt.get("best_epoch",best_epoch))
    return history



````

### 9.0.1 PyTorch model inspection before training

The PyTorch model definition should be read in the same order as the TensorFlow model:

```text
Input → convolutional blocks → pooling → global average pooling → dropout → classifier
```

Although the syntax differs, the tensor flow should be recognizable. This is the central framework-comparison lesson of Lecture 04.



## 9.4 What PyTorch makes visible

The most important contrast with Keras is the training loop.

PyTorch exposes the central cycle explicitly:

```text
optimizer.zero_grad()
        ↓
model(x)
        ↓
criterion(outputs, labels)
        ↓
loss.backward()
        ↓
optimizer.step()
```

This corresponds closely to the learning-cycle diagram from Lecture 04:

**Zero gradients → Forward → Loss → Backward → Update**

The model definition is also explicit through `nn.Module` and `forward()`.

This does not mean PyTorch exposes every low-level operation. Autograd still calculates derivatives automatically, and framework layers still implement convolution, pooling, and batch normalization. The important point is that the programmer sees more of the model data flow and optimization loop than in a one-line `model.fit()` call.

## 10.0.1 How to interpret a result correctly

Suppose one experiment reports 0.89 accuracy and another reports 0.90. The difference is not automatically meaningful. A disciplined interpretation checks the surrounding evidence.

### First: inspect validation behavior

If training loss continues to fall while validation loss rises, the model may be fitting the training data faster than it is improving generalization. Early stopping and checkpoint restoration are intended to reduce this problem.

### Second: inspect class-level behavior

A global accuracy improvement can occur because a large class became easier while a smaller class became worse. Macro F1 and per-class recall make that visible.

### Third: inspect the confusion matrix

Two models can have identical accuracy while making very different mistakes. One model may spread errors across many classes; another may repeatedly confuse the same pair.

### Fourth: inspect probability quality

A wrong prediction at 0.51 confidence and a wrong prediction at 0.99 confidence have the same top-1 outcome but very different implications for reliability. Log loss, confidence distributions, reliability diagrams, and ECE help expose this difference.

### Fifth: inspect computational cost

Extra depth is a trade-off. A small quality improvement may come with a large increase in parameters and runtime. The report should quantify that trade-off instead of describing depth as automatically beneficial.




## 10.1 Reading the results like a technical practitioner

The order of analysis below is recommended because it moves from optimization behavior to generalization, then to class-level errors, and finally to computational cost.

### 1. Training and validation curves

Look for the relationship between training and validation loss. A widening gap can indicate overfitting. A plateau in both curves can indicate that the model has exhausted the useful optimization progress under the current configuration.

### 2. Accuracy together with macro F1

Accuracy describes overall correctness. Macro F1 asks whether the model is performing consistently across classes. For an imbalanced clinical target, this distinction becomes particularly important.

### 3. Per-class recall

Recall tells us what fraction of the true examples of each class were detected. A class with low recall is a concrete error pattern that should be discussed even if the overall score looks strong.

### 4. Confusion matrix

The matrix reveals the direction of each error. In a clinical dataset, confusing `NO` with `>30` is not the same pattern as confusing `NO` with `<30`; in an image dataset, confusing `Pullover` with `Coat` is different from confusing `Sneaker` with `Bag`.

### 5. Confidence and calibration

A reliable classifier should not routinely behave as though it is certain when it is wrong. Calibration diagnostics provide evidence about this behavior.

### 6. Parameter count and time

The final question is computational efficiency. Deeper models naturally tend to require more operations, but the exact relationship depends on where pooling occurs and on implementation details. The table records the actual measured training time rather than assuming the relationship.




# Part I - Error analysis

## 11. Error analysis: from score to explanation

A final metric is a summary, not an explanation. Error analysis examines the remaining mistakes and asks whether they form a meaningful pattern.

### Fashion-MNIST

The notebook visualizes misclassified clothing images together with the true label, predicted label, and confidence. Repeated confusions between visually similar upper-body categories can indicate that the task contains genuine visual ambiguity rather than a coding error.

### CIFAR-100

The 100-class setting benefits from two levels of analysis:

1. the full numeric confusion matrix, which preserves all evidence;
2. the top confusion pairs, which provide a readable ranking of the most frequent class-to-class errors.

The correct interpretation is not "the model failed because the matrix is large." A 100 x 100 matrix is large because the task itself contains 100 classes.

### Diabetes

The clinical dataset is analyzed through target-specific precision, recall, F1, confusion counts, confidence, and calibration. The notebook avoids claiming that a model is clinically useful; this assignment evaluates classification behavior on the supplied dataset, not a clinical deployment system.

### High-confidence errors

A particularly useful subset is the set of wrong predictions with high confidence. These examples can indicate insufficient representation, ambiguous samples, or a mismatch between the learned decision boundary and the data distribution.

Error analysis should remain descriptive. It is evidence about model behavior, not a license to invent causes that the dataset cannot support.




### 11.1 Error-analysis checklist

For every model, ask:

1. Which classes are hardest?
2. Are the hardest classes visually or statistically similar?
3. Are errors concentrated in one direction?
4. Does the deeper model change the error pattern or only the aggregate score?
5. Are wrong predictions unusually confident?
6. Does the framework change the class-level behavior even when the architecture is conceptually the same?

These questions force the report to explain model behavior instead of merely copying the final accuracy value.

# 13. Technical conclusion

## 13.1 What A5 demonstrates at the data level

The three datasets require different data reasoning before CNN training begins.

Fashion-MNIST is a compact grayscale image problem where the main preparation concerns shape, numeric range, normalization, and class integrity.

CIFAR-100 is an RGB image problem with a much larger label space. The main additional challenge is the number of classes and the resulting need for richer evaluation, especially top-5 accuracy and class-level confusion analysis.

Diabetes 130-US Hospitals is a large heterogeneous clinical table. Its main preparation challenges are missing values, mixed data types, repeated patients, target imbalance, and leakage prevention.

This contrast is important because the quality of a neural-network experiment is influenced by data preparation decisions long before the first gradient update occurs.

## 13.2 What A5 demonstrates at the architecture level

CNN3 and CNN5 share the same fundamental operations but differ in depth and representational capacity.

The added convolutional stages in CNN5 allow the network to apply more successive transformations before classification. That can be useful when the dataset contains more complex visual structure, but greater depth also increases computation and can change optimization behavior.

The correct conclusion therefore comes from the measured evidence:

```text
Depth change
    ↓
Parameter count
    ↓
Training dynamics
    ↓
Validation behavior
    ↓
Test metrics
    ↓
Confusion pattern
    ↓
Confidence / calibration
    ↓
Runtime trade-off
```

## 13.3 What A5 demonstrates about TensorFlow and PyTorch

Following Lecture 04, the two frameworks should not be interpreted as two different mathematical learning methods.

TensorFlow/Keras raises the abstraction level through reusable layers and high-level training APIs such as `compile()` and `fit()`.

PyTorch keeps the model structure explicit through `nn.Module` and `forward()`, while exposing the optimization cycle through `zero_grad()`, `loss.backward()`, and `optimizer.step()`.

The computational idea remains:

```text
Input
  ↓
Transformations
  ↓
Prediction
  ↓
Loss
  ↓
Gradient
  ↓
Parameter update
```

The programming abstraction changes, not the underlying learning objective.

## 13.4 What should and should not be concluded from the Diabetes Conv1D experiment

The A5 requirement forces a CNN experiment on Diabetes, so Conv1D provides a consistent way to study CNN mechanics on the third dataset. However, the report should retain the limitation emphasized in Lecture 04: ordinary tabular columns do not naturally form a spatial neighborhood like pixels in an image.

Therefore the Diabetes Conv1D result is best interpreted as an architectural experiment under the assignment constraint. It should not be presented as evidence that Conv1D is universally the correct architecture for clinical tabular data.

## 13.5 Final reporting principle

The strongest technical report is not the one that contains the largest number of decimals. It is the one where every important number is connected to:

- a clear data pipeline;
- a documented architecture;
- controlled experimental settings;
- appropriate metrics;
- visual evidence;
- an explanation of errors and trade-offs.

The final conclusion of A5 should therefore be written after the full run and should refer directly to the observed results rather than to expectations established before training.

## 14. Lecture 04 framework comparison table

The following table maps the same CNN concept across the two required frameworks and the explicit NumPy fundamentals demonstrated earlier.

| Concept | NumPy fundamentals | TensorFlow/Keras | PyTorch |
|---|---|---|---|
| Convolution | explicit numerical function | `Conv2D` / `Conv1D` | `nn.Conv2d` / `nn.Conv1d` |
| Activation | `relu()` | `ReLU()` or activation layer | `nn.ReLU()` |
| Pooling | `max_pool2d()` example | `MaxPooling2D` / `MaxPooling1D` | `nn.MaxPool2d` / `nn.MaxPool1d` |
| Global aggregation | explicit arithmetic concept | `GlobalAveragePooling...` | `AdaptiveAvgPool...` |
| Classifier | matrix multiplication concept | `Dense` | `nn.Linear` |
| Loss | explicit cross-entropy demonstration | `SparseCategoricalCrossentropy` | `CrossEntropyLoss` |
| Gradient | conceptually discussed | automatic differentiation | autograd |
| Training loop | manual concept | largely encapsulated by `fit()` | explicit loop |
| Model structure | explicit operations | layer composition | `nn.Module` + `forward()` |

### Main lesson

Different function names do not imply different machine-learning concepts. The important skill is to identify the role of each component and the tensor transformation it performs.
