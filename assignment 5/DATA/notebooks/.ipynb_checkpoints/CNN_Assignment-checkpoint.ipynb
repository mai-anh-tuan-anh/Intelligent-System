# Part B - CNN fundamentals

## 3. Các phép toán cơ bản

Một CNN có thể được hiểu như một chuỗi hàm:

```text
input image
    -> convolution
    -> activation
    -> pooling
    -> deeper feature maps
    -> global pooling / flatten
    -> classifier
```

Ở mức minh họa, ta viết các hàm bằng NumPy để nhìn thấy phép tính. Framework sẽ dùng implementation tối ưu hơn nhiều.


````python
def relu(x):
    """Đưa giá trị âm về 0."""
    return np.maximum(0, x)


def softmax(logits):
    """Đổi logits thành xác suất theo chiều cuối."""
    shifted = logits - np.max(logits, axis=-1, keepdims=True)
    exp_x = np.exp(shifted)
    return exp_x / exp_x.sum(axis=-1, keepdims=True)


def sparse_cross_entropy(logits, y_true):
    """Cross-entropy cho nhãn integer."""
    probs = softmax(logits)
    p_true = probs[np.arange(len(y_true)), y_true]
    return float(-np.log(np.clip(p_true, 1e-12, 1.0)).mean())

print("ReLU:", relu(np.array([-2, -1, 0, 1, 2])))
logits = np.array([[2.0, 1.0, 0.1]])
print("Softmax:", softmax(logits))
print("Cross-entropy:", sparse_cross_entropy(logits, np.array([0])))
````

````python
def conv2d_single_channel(image, kernel, stride=1, padding=0):
    """Convolution 2D cho 1 ảnh một channel và 1 kernel."""
    image = np.asarray(image, dtype=np.float32)
    kernel = np.asarray(kernel, dtype=np.float32)
    if padding:
        image = np.pad(image, ((padding, padding), (padding, padding)))

    kh, kw = kernel.shape
    out_h = (image.shape[0] - kh) // stride + 1
    out_w = (image.shape[1] - kw) // stride + 1
    out = np.zeros((out_h, out_w), dtype=np.float32)

    for i in range(out_h):
        for j in range(out_w):
            region = image[i*stride:i*stride+kh, j*stride:j*stride+kw]
            out[i, j] = np.sum(region * kernel)
    return out


def max_pool2d(feature_map, pool_size=2, stride=2):
    """Max pooling: lấy giá trị lớn nhất trong mỗi cửa sổ."""
    h, w = feature_map.shape
    out_h = (h - pool_size) // stride + 1
    out_w = (w - pool_size) // stride + 1
    out = np.zeros((out_h, out_w), dtype=np.float32)

    for i in range(out_h):
        for j in range(out_w):
            region = feature_map[i*stride:i*stride+pool_size, j*stride:j*stride+pool_size]
            out[i, j] = np.max(region)
    return out

example_image = np.arange(25, dtype=np.float32).reshape(5, 5)
example_kernel = np.array([[1, 0, -1], [1, 0, -1], [1, 0, -1]], dtype=np.float32)
feature = conv2d_single_channel(example_image, example_kernel)
pooled = max_pool2d(feature)
print("Convolution output:\n", feature)
print("Pooled output:\n", pooled)
````

## 3.1 Vì sao so sánh 3 và 5 convolutional layers?

Trong bài này, “3 layers” và “5 layers” được hiểu rõ là **3 hoặc 5 convolutional layers**. Mỗi convolutional layer có thể đi cùng Batch Normalization, ReLU và một số bước pooling.

Mô hình sâu hơn có thêm cơ hội xây dựng feature phức tạp từ các feature đơn giản. Nhưng depth cũng làm tăng số phép tính và có thể làm overfit. Vì vậy A5 phải trả lời bằng dữ liệu thực nghiệm: validation curve, macro F1, per-class metrics, confusion matrix, calibration và training cost.


# Part C - Common audit and visualization utilities


````python
def plot_class_distribution(labels, class_names=None, title="Class distribution"):
    values, counts = np.unique(labels, return_counts=True)
    names = [class_names[int(v)] if class_names is not None else str(v) for v in values]
    plt.figure(figsize=(12, 5))
    bars = plt.bar(names, counts)
    plt.title(title)
    plt.xlabel("Class")
    plt.ylabel("Count")
    plt.xticks(rotation=45, ha="right")
    for bar, count in zip(bars, counts):
        plt.text(bar.get_x()+bar.get_width()/2, bar.get_height(), f"{count:,}", ha="center", va="bottom", fontsize=8)
    plt.tight_layout(); plt.show()


def plot_image_grid(images, labels, class_names, n=12, title="Samples"):
    n = min(n, len(images))
    cols = 4
    rows = math.ceil(n / cols)
    plt.figure(figsize=(12, 3*rows))
    for i in range(n):
        ax = plt.subplot(rows, cols, i+1)
        ax.imshow(images[i], cmap="gray" if images[i].ndim == 2 else None)
        ax.set_title(class_names[int(labels[i])])
        ax.axis("off")
    plt.suptitle(title); plt.tight_layout(); plt.show()


def plot_image_statistics(images, title):
    flat = images.astype(np.float32).reshape(-1)
    plt.figure(figsize=(9, 4))
    plt.hist(flat, bins=50)
    plt.title(title + " - pixel distribution")
    plt.xlabel("Pixel value"); plt.ylabel("Frequency")
    plt.tight_layout(); plt.show()

    mean_img = images.astype(np.float32).mean(axis=0)
    std_img = images.astype(np.float32).std(axis=0)
    if mean_img.ndim == 3:
        mean_img = mean_img.mean(axis=-1)
        std_img = std_img.mean(axis=-1)
    fig, ax = plt.subplots(1, 2, figsize=(10, 4))
    ax[0].imshow(mean_img, cmap="gray"); ax[0].set_title("Mean image"); ax[0].axis("off")
    ax[1].imshow(std_img, cmap="magma"); ax[1].set_title("Pixel std map"); ax[1].axis("off")
    plt.suptitle(title); plt.tight_layout(); plt.show()


def report_image_integrity(images, labels, name, n_classes):
    x = images.astype(np.float32)
    print(f"[{name}]")
    print("shape:", images.shape, "dtype:", images.dtype)
    print("NaN:", int(np.isnan(x).sum()), "Inf:", int(np.isinf(x).sum()))
    print("pixel range:", float(x.min()), "to", float(x.max()))
    print("invalid labels:", int(((labels < 0) | (labels >= n_classes)).sum()))


def split_image_train_val(X, y, val_size=0.10, seed=42, train_limit=None, val_limit=None):
    idx = np.arange(len(y))
    train_idx, val_idx = train_test_split(idx, test_size=val_size, stratify=y, random_state=seed)
    if train_limit is not None and train_limit < len(train_idx):
        train_idx, _ = train_test_split(train_idx, train_size=train_limit, stratify=y[train_idx], random_state=seed)
    if val_limit is not None and val_limit < len(val_idx):
        val_idx, _ = train_test_split(val_idx, train_size=val_limit, stratify=y[val_idx], random_state=seed)
    return np.sort(train_idx), np.sort(val_idx)
````

## 6.1 Cleaning strategy

Đây là phần quan trọng nhất của Diabetes vì dữ liệu nhiều categorical feature và có sentinel `?`.

### Identifier

`encounter_id` và `patient_nbr` là identifier nên được bỏ khỏi feature matrix. Model không cần học giá trị định danh.

### Missing data

Sau khi đổi `?` thành `NaN`, ta không xóa hàng loạt. Các cột có missing rate > 80% được loại bỏ. Phần còn lại:

- numeric -> median imputation;
- categorical -> most-frequent imputation.

### Encoding

Categorical -> One-Hot Encoding với `handle_unknown="ignore"`.
Numeric -> StandardScaler.

Quan trọng: **preprocessor chỉ fit trên training split**. Validation/test chỉ được transform, tránh data leakage.

### Class imbalance

Loss sử dụng class weights tính từ training set. Khi báo cáo cũng dùng macro F1 và per-class recall để không bị đánh lừa bởi class lớn nhất.


````python
def prepare_diabetes(df, seed=42, run_mode="full"):
    df = df.replace("?", np.nan).copy()
    y = df["readmitted"].map({"NO": 0, ">30": 1, "<30": 2}).astype(int).to_numpy()

    # Drop identifiers before splitting; they are not model features.
    X = df.drop(columns=["readmitted", "encounter_id", "patient_nbr"], errors="ignore")

    idx = np.arange(len(y))
    train_idx, temp_idx = train_test_split(idx, test_size=0.20, stratify=y, random_state=seed)
    val_idx, test_idx = train_test_split(temp_idx, test_size=0.50, stratify=y[temp_idx], random_state=seed)

    if run_mode == "pilot":
        train_limit = min(PILOT_LIMITS["diabetes_train"], len(train_idx))
        val_limit = min(PILOT_LIMITS["diabetes_val"], len(val_idx))
        train_idx, _ = train_test_split(train_idx, train_size=train_limit, stratify=y[train_idx], random_state=seed)
        val_idx, _ = train_test_split(val_idx, train_size=val_limit, stratify=y[val_idx], random_state=seed)

    X_train_df, X_val_df, X_test_df = X.iloc[train_idx].copy(), X.iloc[val_idx].copy(), X.iloc[test_idx].copy()
    y_train, y_val, y_test = y[train_idx], y[val_idx], y[test_idx]

    # Learn the high-missingness decision from the training split only.
    missing_rate = X_train_df.isna().mean()
    high_missing = missing_rate[missing_rate > 0.80].index.tolist()
    X_train_df = X_train_df.drop(columns=high_missing, errors="ignore")
    X_val_df = X_val_df.drop(columns=high_missing, errors="ignore")
    X_test_df = X_test_df.drop(columns=high_missing, errors="ignore")

    numeric_cols = X_train_df.select_dtypes(include=[np.number]).columns.tolist()
    categorical_cols = [c for c in X_train_df.columns if c not in numeric_cols]

    # These are categorical codes, not continuous measurements.
    for col in ["admission_type_id", "discharge_disposition_id", "admission_source_id"]:
        if col in numeric_cols:
            numeric_cols.remove(col)
            categorical_cols.append(col)

    try:
        ohe = OneHotEncoder(handle_unknown="ignore", sparse_output=False, dtype=np.float32)
    except TypeError:
        ohe = OneHotEncoder(handle_unknown="ignore", sparse=False, dtype=np.float32)

    num_pipe = Pipeline([
        ("imputer", SimpleImputer(strategy="median")),
        ("scaler", StandardScaler()),
    ])
    cat_pipe = Pipeline([
        ("imputer", SimpleImputer(strategy="most_frequent")),
        ("onehot", ohe),
    ])

    preprocessor = ColumnTransformer([
        ("num", num_pipe, numeric_cols),
        ("cat", cat_pipe, categorical_cols),
    ])

    # Fit the preprocessing pipeline on train only.
    X_train = preprocessor.fit_transform(X_train_df).astype(np.float32)
    X_val = preprocessor.transform(X_val_df).astype(np.float32)
    X_test = preprocessor.transform(X_test_df).astype(np.float32)

    # Conv1D expects (samples, sequence_length, channels).
    X_train = X_train[..., None]
    X_val = X_val[..., None]
    X_test = X_test[..., None]

    classes = np.unique(y_train)
    weights = compute_class_weight(class_weight="balanced", classes=classes, y=y_train)
    class_weight = {int(c): float(w) for c, w in zip(classes, weights)}

    return {
        "X_train": X_train, "y_train": y_train,
        "X_val": X_val, "y_val": y_val,
        "X_test": X_test, "y_test": y_test,
        "preprocessor": preprocessor,
        "class_weight": class_weight,
        "dropped_columns": ["encounter_id", "patient_nbr"] + high_missing,
        "duplicate_count_raw": int(df.duplicated().sum()),
        "feature_count": X_train.shape[1],
        "numeric_cols": numeric_cols,
        "categorical_cols": categorical_cols,
    }

diabetes_data = prepare_diabetes(diabetes_raw, SEED, RUN_MODE)
X_diab_tr, y_diab_tr = diabetes_data["X_train"], diabetes_data["y_train"]
X_diab_val, y_diab_val = diabetes_data["X_val"], diabetes_data["y_val"]
X_diab_test, y_diab_test = diabetes_data["X_test"], diabetes_data["y_test"]
DIABETES_NAMES = ["NO", ">30", "<30"]

print("train:", X_diab_tr.shape, "val:", X_diab_val.shape, "test:", X_diab_test.shape)
print("Encoded features:", diabetes_data["feature_count"])
print("Dropped columns:", diabetes_data["dropped_columns"])
print("Class weights:", diabetes_data["class_weight"])
plot_class_distribution(y_diab_tr, DIABETES_NAMES, "Diabetes training class distribution")
````

# Part G - Shared evaluation and visualization

Ta dùng một bộ hàm đánh giá chung cho tất cả 12 experiments. Điều này tránh tình trạng TensorFlow dùng một metric, PyTorch dùng metric khác rồi so sánh không công bằng.


````python
def top_k_accuracy(y_true, probs, k=5):
    k = min(k, probs.shape[1])
    topk = np.argsort(probs, axis=1)[:, -k:]
    return float(np.mean([y in row for y, row in zip(y_true, topk)]))


def ece_score(y_true, probs, n_bins=10):
    conf = probs.max(axis=1)
    pred = probs.argmax(axis=1)
    correct = pred == y_true
    edges = np.linspace(0, 1, n_bins + 1)
    ece = 0.0
    for i in range(n_bins):
        lo, hi = edges[i], edges[i+1]
        mask = (conf >= lo) & ((conf < hi) if i < n_bins-1 else (conf <= hi))
        if mask.any():
            ece += mask.mean() * abs(correct[mask].mean() - conf[mask].mean())
    return float(ece)


def evaluate_predictions(y_true, probs, class_names):
    preds = probs.argmax(axis=1)
    p_m, r_m, f1_m, _ = precision_recall_fscore_support(y_true, preds, average="macro", zero_division=0)
    p_w, r_w, f1_w, _ = precision_recall_fscore_support(y_true, preds, average="weighted", zero_division=0)
    out = {
        "accuracy": float(accuracy_score(y_true, preds)),
        "precision_macro": float(p_m),
        "recall_macro": float(r_m),
        "f1_macro": float(f1_m),
        "precision_weighted": float(p_w),
        "recall_weighted": float(r_w),
        "f1_weighted": float(f1_w),
        "log_loss": float(log_loss(y_true, probs, labels=np.arange(len(class_names)))),
        "mean_confidence": float(probs.max(axis=1).mean()),
        "ece": ece_score(y_true, probs),
    }
    if probs.shape[1] >= 5:
        out["top5_accuracy"] = top_k_accuracy(y_true, probs, 5)
    return out, preds


def plot_history(history, title):
    epochs = np.arange(1, len(history["loss"])+1)
    fig, ax = plt.subplots(1, 2, figsize=(12, 4))
    ax[0].plot(epochs, history["loss"], marker="o", label="train")
    ax[0].plot(epochs, history["val_loss"], marker="o", label="val")
    ax[0].set_title(title + " - loss"); ax[0].set_xlabel("Epoch"); ax[0].set_ylabel("Loss"); ax[0].legend()
    ax[1].plot(epochs, history["accuracy"], marker="o", label="train")
    ax[1].plot(epochs, history["val_accuracy"], marker="o", label="val")
    ax[1].set_title(title + " - accuracy"); ax[1].set_xlabel("Epoch"); ax[1].set_ylabel("Accuracy"); ax[1].legend()
    plt.tight_layout(); plt.show()


def plot_confusion(y_true, preds, class_names, title, dataset_key):
    cm = confusion_matrix(y_true, preds)
    n = len(class_names)
    fig, ax = plt.subplots(figsize=(8 if n <= 20 else 12, 8 if n <= 20 else 12))
    im = ax.imshow(cm, cmap="Blues")
    plt.colorbar(im, ax=ax)
    ax.set_xlabel("Predicted"); ax.set_ylabel("True"); ax.set_title(title + " - count confusion matrix")
    if n <= 20:
        ax.set_xticks(range(n)); ax.set_yticks(range(n))
        ax.set_xticklabels(class_names, rotation=45, ha="right", fontsize=8)
        ax.set_yticklabels(class_names, fontsize=8)
        threshold = cm.max() * 0.5 if cm.size else 0
        for i in range(n):
            for j in range(n):
                ax.text(j, i, str(cm[i,j]), ha="center", va="center", fontsize=7,
                        color="white" if cm[i,j] > threshold else "black")
    else:
        ax.set_xticks(range(n)); ax.set_yticks(range(n))
        ax.set_xticklabels(range(n), fontsize=5); ax.set_yticklabels(range(n), fontsize=5)
    fig.savefig(FIGURES_DIR / f"{dataset_key}_{title.replace(' ', '_').lower()}_cm.png", dpi=160, bbox_inches="tight")
    plt.show()

    if n > 20:
        off = cm.copy(); np.fill_diagonal(off, 0)
        important = np.argsort(off.sum(axis=1))[-20:]
        sub = cm[np.ix_(important, important)]
        names = [class_names[i] for i in important]
        fig, ax = plt.subplots(figsize=(12, 10))
        im = ax.imshow(sub, cmap="Blues"); plt.colorbar(im, ax=ax)
        ax.set_xticks(range(20)); ax.set_yticks(range(20))
        ax.set_xticklabels(names, rotation=60, ha="right", fontsize=7); ax.set_yticklabels(names, fontsize=7)
        for i in range(20):
            for j in range(20):
                ax.text(j, i, str(sub[i,j]), ha="center", va="center", fontsize=6,
                        color="white" if sub[i,j] > sub.max()*0.5 else "black")
        ax.set_title(title + " - annotated top confusion subset")
        plt.tight_layout(); plt.show()
    return cm


def plot_per_class_metrics(y_true, preds, class_names, title):
    report = classification_report(y_true, preds, target_names=class_names, output_dict=True, zero_division=0)
    df = pd.DataFrame(report).T.iloc[:len(class_names)]
    ax = df[["precision", "recall", "f1-score"]].plot(kind="bar", figsize=(12, max(5, len(class_names)*0.18)))
    ax.set_ylim(0, 1.05); ax.set_title(title + " - per-class metrics"); ax.set_ylabel("Score")
    if len(class_names) <= 20:
        ax.tick_params(axis="x", rotation=35)
    else:
        ax.set_xticks([])
    plt.tight_layout(); plt.show()
    return df


def plot_confidence(y_true, probs, title):
    pred = probs.argmax(axis=1)
    conf = probs.max(axis=1)
    correct = pred == y_true
    plt.figure(figsize=(9,4))
    plt.hist(conf[correct], bins=20, alpha=0.7, label="Correct")
    plt.hist(conf[~correct], bins=20, alpha=0.7, label="Incorrect")
    plt.xlabel("Confidence"); plt.ylabel("Count"); plt.title(title + " - confidence distribution"); plt.legend(); plt.tight_layout(); plt.show()


def plot_reliability(y_true, probs, title, n_bins=10):
    conf = probs.max(axis=1); pred = probs.argmax(axis=1); correct = pred == y_true
    edges = np.linspace(0, 1, n_bins+1); bin_c=[]; bin_a=[]
    for i in range(n_bins):
        mask = (conf >= edges[i]) & ((conf < edges[i+1]) if i < n_bins-1 else (conf <= edges[i+1]))
        if mask.any():
            bin_c.append(conf[mask].mean()); bin_a.append(correct[mask].mean())
    plt.figure(figsize=(6,6)); plt.plot([0,1],[0,1], "--", label="Perfect calibration"); plt.plot(bin_c, bin_a, "o-", label="Model")
    plt.xlabel("Mean confidence"); plt.ylabel("Accuracy"); plt.title(title + " - reliability diagram"); plt.legend(); plt.tight_layout(); plt.show()


def plot_misclassified(images, y_true, preds, probs, class_names, title, n=12):
    wrong = np.where(y_true != preds)[0][:n]
    if len(wrong) == 0:
        print("No misclassified examples."); return
    cols = 4; rows = math.ceil(len(wrong)/cols)
    plt.figure(figsize=(12,3*rows))
    for k, idx in enumerate(wrong):
        ax = plt.subplot(rows, cols, k+1)
        ax.imshow(images[idx], cmap="gray" if images[idx].ndim == 2 else None)
        ax.set_title(f"True: {class_names[y_true[idx]]}\nPred: {class_names[preds[idx]]} | {probs[idx].max():.3f}")
        ax.axis("off")
    plt.suptitle(title + " - error examples"); plt.tight_layout(); plt.show()


def top_confusion_pairs(y_true, preds, class_names, top_n=15):
    cm = confusion_matrix(y_true, preds)
    pairs=[]
    for i in range(len(class_names)):
        for j in range(len(class_names)):
            if i != j and cm[i,j] > 0:
                pairs.append((int(cm[i,j]), class_names[i], class_names[j]))
    pairs.sort(reverse=True)
    return pd.DataFrame(pairs[:top_n], columns=["count","true_class","predicted_class"])


def save_experiment(dataset_key, framework, depth, y_true, probs, metrics):
    prefix = f"{dataset_key}_{framework}_cnn{depth}"
    np.savez_compressed(PREDICTIONS_DIR / f"{prefix}_predictions.npz", y_true=y_true, y_pred=probs.argmax(1), probabilities=probs)
    with open(METRICS_DIR / f"{prefix}.json", "w", encoding="utf-8") as f:
        json.dump(metrics, f, indent=2)
    pd.DataFrame(confusion_matrix(y_true, probs.argmax(1))).to_csv(METRICS_DIR / f"{prefix}_confusion_matrix.csv", index=False)


def evaluate_and_visualize(dataset_key, framework, depth, y_true, probs, class_names, images=None, history=None):
    metrics, preds = evaluate_predictions(y_true, probs, class_names)
    metrics.update({"dataset": dataset_key, "framework": framework, "conv_layers": depth})
    print(f"\n{dataset_key} | {framework} | CNN{depth}")
    display(pd.DataFrame([metrics]).T.rename(columns={0:"value"}))
    if history is not None:
        plot_history(history, f"{dataset_key} {framework} CNN{depth}")
    plot_confusion(y_true, preds, class_names, f"{dataset_key} {framework} CNN{depth}", dataset_key)
    print("Per-class metrics:")
    display(plot_per_class_metrics(y_true, preds, class_names, f"{dataset_key} {framework} CNN{depth}"))
    plot_confidence(y_true, probs, f"{dataset_key} {framework} CNN{depth}")
    plot_reliability(y_true, probs, f"{dataset_key} {framework} CNN{depth}")
    if images is not None:
        plot_misclassified(images, y_true, preds, probs, class_names, f"{dataset_key} {framework} CNN{depth}")
    print("Top confusion pairs:")
    display(top_confusion_pairs(y_true, preds, class_names))
    save_experiment(dataset_key, framework, depth, y_true, probs, metrics)
    return metrics, preds
````

# Part H - TensorFlow / Keras CNN

## 7. Kiến trúc

Để phép so sánh rõ ràng, 3-layer và 5-layer dùng cùng kiểu block:

```text
Conv -> BatchNorm -> ReLU -> (Pool)
```

Sau block cuối dùng GlobalAveragePooling rồi Dropout và Dense logits. Với image dùng Conv2D; với Diabetes dùng Conv1D.


````python
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers


def build_tf_cnn(input_shape, num_classes, depth=3, task_type="image"):
    if depth not in (3,5):
        raise ValueError("depth must be 3 or 5")
    inputs = keras.Input(shape=input_shape)
    x = inputs

    if task_type == "image":
        x = layers.RandomFlip("horizontal")(x)
        x = layers.RandomRotation(0.03)(x)
        filters = [32,64,128,128,160] if depth == 5 else [32,64,128]
        pool_after = {0,1} if depth == 3 else {0,1,3}
        for i in range(depth):
            x = layers.Conv2D(filters[i], 3, padding="same", use_bias=False)(x)
            x = layers.BatchNormalization()(x)
            x = layers.ReLU()(x)
            if i in pool_after:
                x = layers.MaxPooling2D(2)(x)
        x = layers.GlobalAveragePooling2D()(x)
    elif task_type == "tabular":
        filters = [32,64,128,128,160] if depth == 5 else [32,64,128]
        pool_after = {0,1} if depth == 3 else {0,1,3}
        for i in range(depth):
            x = layers.Conv1D(filters[i], 3, padding="same", use_bias=False)(x)
            x = layers.BatchNormalization()(x)
            x = layers.ReLU()(x)
            if i in pool_after:
                x = layers.MaxPooling1D(2)(x)
        x = layers.GlobalAveragePooling1D()(x)
    else:
        raise ValueError("task_type must be image or tabular")

    x = layers.Dropout(0.30)(x)
    outputs = layers.Dense(num_classes)(x)
    return keras.Model(inputs, outputs)


def train_tf(model, X_train, y_train, X_val, y_val, class_weight, model_path, log_dir):
    model.compile(
        optimizer=keras.optimizers.Adam(learning_rate=LEARNING_RATE),
        loss=keras.losses.SparseCategoricalCrossentropy(from_logits=True),
        metrics=[keras.metrics.SparseCategoricalAccuracy(name="accuracy")],
    )
    callbacks = [
        keras.callbacks.EarlyStopping(monitor="val_loss", patience=3, restore_best_weights=True),
        keras.callbacks.ReduceLROnPlateau(monitor="val_loss", factor=0.5, patience=2, min_lr=1e-6, verbose=1),
        keras.callbacks.ModelCheckpoint(model_path, monitor="val_loss", save_best_only=True),
        keras.callbacks.TensorBoard(log_dir=log_dir),
    ]
    start = time.perf_counter()
    hist = model.fit(
        X_train, y_train,
        validation_data=(X_val, y_val),
        epochs=EPOCHS,
        batch_size=BATCH_SIZE,
        shuffle=True,
        class_weight=class_weight,
        callbacks=callbacks,
        verbose=2,
    )
    out = {k:[float(v) for v in vals] for k, vals in hist.history.items()}
    out["training_seconds"] = time.perf_counter() - start
    return out


def predict_tf(model, X):
    logits = model.predict(X, batch_size=TEST_BATCH_SIZE, verbose=0)
    probs = tf.nn.softmax(logits, axis=1).numpy()
    return logits, probs
````

# Part I - PyTorch CNN

PyTorch sử dụng Dataset và DataLoader để tách dữ liệu khỏi training loop. Với Windows, đặt `num_workers=0` giúp pipeline đơn giản và tránh một số vấn đề multiprocessing thường gặp.


````python
import torch
from torch import nn
from torch.utils.data import Dataset, DataLoader

DEVICE = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print("Device:", DEVICE)

try:
    from torchvision import transforms
    HAS_TORCHVISION = True
except Exception as exc:
    HAS_TORCHVISION = False
    print("torchvision warning:", exc)


class NumpyImageDataset(Dataset):
    def __init__(self, X, y, train=False):
        self.X = X
        self.y = y.astype(np.int64)
        self.augment = None
        if HAS_TORCHVISION and train:
            self.augment = transforms.Compose([
                transforms.RandomHorizontalFlip(0.5),
                transforms.RandomCrop(X.shape[1], padding=2),
            ])

    def __len__(self):
        return len(self.y)

    def __getitem__(self, idx):
        img = self.X[idx]
        if img.ndim == 2:
            img = img[...,None]
        tensor = torch.from_numpy(img.transpose(2,0,1)).float()
        if self.augment is not None:
            tensor = self.augment(tensor)
        return tensor, torch.tensor(self.y[idx], dtype=torch.long)


class NumpyTabularDataset(Dataset):
    def __init__(self, X, y):
        self.X = torch.from_numpy(X).float()
        self.y = torch.from_numpy(y.astype(np.int64)).long()
    def __len__(self): return len(self.y)
    def __getitem__(self, idx): return self.X[idx], self.y[idx]


class ImageCNN(nn.Module):
    def __init__(self, in_channels, num_classes, depth=3):
        super().__init__()
        filters = [32,64,128,128,160] if depth == 5 else [32,64,128]
        pool_after = {0,1} if depth == 3 else {0,1,3}
        blocks=[]
        for i in range(depth):
            blocks += [
                nn.Conv2d(in_channels if i==0 else filters[i-1], filters[i], 3, padding=1, bias=False),
                nn.BatchNorm2d(filters[i]),
                nn.ReLU(inplace=True),
            ]
            if i in pool_after:
                blocks.append(nn.MaxPool2d(2))
        self.features = nn.Sequential(*blocks)
        self.pool = nn.AdaptiveAvgPool2d(1)
        self.dropout = nn.Dropout(0.30)
        self.classifier = nn.Linear(filters[-1], num_classes)

    def forward(self, x):
        x = self.features(x)
        x = self.pool(x).flatten(1)
        return self.classifier(self.dropout(x))


class TabularCNN(nn.Module):
    def __init__(self, num_classes, depth=3):
        super().__init__()
        filters = [32,64,128,128,160] if depth == 5 else [32,64,128]
        pool_after = {0,1} if depth == 3 else {0,1,3}
        blocks=[]
        for i in range(depth):
            blocks += [
                nn.Conv1d(1 if i==0 else filters[i-1], filters[i], 3, padding=1, bias=False),
                nn.BatchNorm1d(filters[i]),
                nn.ReLU(inplace=True),
            ]
            if i in pool_after:
                blocks.append(nn.MaxPool1d(2))
        self.features = nn.Sequential(*blocks)
        self.pool = nn.AdaptiveAvgPool1d(1)
        self.dropout = nn.Dropout(0.30)
        self.classifier = nn.Linear(filters[-1], num_classes)

    def forward(self, x):
        x = self.features(x)
        x = self.pool(x).flatten(1)
        return self.classifier(self.dropout(x))


def loader(ds, shuffle=False):
    return DataLoader(ds, batch_size=BATCH_SIZE, shuffle=shuffle, num_workers=0, pin_memory=torch.cuda.is_available())


def val_loss_acc(model, dl, criterion):
    model.eval(); total_loss=0.0; total=0; correct=0
    with torch.no_grad():
        for xb,yb in dl:
            xb,yb=xb.to(DEVICE),yb.to(DEVICE)
            logits=model(xb); loss=criterion(logits,yb)
            total_loss += loss.item()*yb.size(0); total += yb.size(0); correct += (logits.argmax(1)==yb).sum().item()
    return total_loss/max(total,1), correct/max(total,1)


def predict_torch(model, dl):
    model.eval(); logits_all=[]; y_all=[]
    with torch.no_grad():
        for xb,yb in dl:
            logits_all.append(model(xb.to(DEVICE)).cpu().numpy()); y_all.append(yb.numpy())
    logits=np.concatenate(logits_all); y_true=np.concatenate(y_all)
    p=np.exp(logits-logits.max(axis=1,keepdims=True)); p/=p.sum(axis=1,keepdims=True)
    return y_true, logits, p


def train_torch(model, train_dl, val_dl, class_weight, path):
    weights=torch.tensor([class_weight.get(i,1.0) for i in range(len(class_weight))], dtype=torch.float32, device=DEVICE)
    criterion=nn.CrossEntropyLoss(weight=weights)
    optimizer=torch.optim.AdamW(model.parameters(), lr=LEARNING_RATE, weight_decay=1e-4)
    scheduler=torch.optim.lr_scheduler.ReduceLROnPlateau(optimizer, mode="min", factor=0.5, patience=2, min_lr=1e-6)
    history={"loss":[],"val_loss":[],"accuracy":[],"val_accuracy":[]}
    best=float("inf"); bad=0; start=time.perf_counter()

    for epoch in range(1,EPOCHS+1):
        model.train(); running=0.0; total=0; correct=0
        for xb,yb in train_dl:
            xb,yb=xb.to(DEVICE),yb.to(DEVICE)
            optimizer.zero_grad(set_to_none=True)
            logits=model(xb); loss=criterion(logits,yb); loss.backward()
            torch.nn.utils.clip_grad_norm_(model.parameters(),5.0)
            optimizer.step()
            running += loss.item()*yb.size(0); total += yb.size(0); correct += (logits.argmax(1)==yb).sum().item()
        tr_loss=running/max(total,1); tr_acc=correct/max(total,1)
        vl_loss,vl_acc=val_loss_acc(model,val_dl,criterion); scheduler.step(vl_loss)
        for k,v in [("loss",tr_loss),("accuracy",tr_acc),("val_loss",vl_loss),("val_accuracy",vl_acc)]: history[k].append(float(v))
        print(f"Epoch {epoch:02d} | train loss {tr_loss:.4f} acc {tr_acc:.4f} | val loss {vl_loss:.4f} acc {vl_acc:.4f}")
        if vl_loss < best-1e-5:
            best=vl_loss; bad=0
            torch.save({"model_state_dict":model.state_dict(),"best_val_loss":best}, path)
        else:
            bad += 1
            if bad >= 3:
                print("Early stopping."); break

    ckpt=torch.load(path, map_location=DEVICE)
    model.load_state_dict(ckpt["model_state_dict"])
    history["training_seconds"]=time.perf_counter()-start
    return history
````

## 9.1 Cách đọc kết quả chuyên gia

Không kết luận chỉ từ một cột. Với mỗi dataset, hãy đọc theo thứ tự:

1. **Validation curves:** train tiếp tục giảm nhưng validation tăng -> dấu hiệu overfitting.
2. **Macro F1:** model có cải thiện đều giữa các class hay chỉ tối ưu class lớn?
3. **Per-class recall:** class nào vẫn khó?
4. **Confusion matrix:** model nhầm những cặp nào?
5. **Confidence/ECE:** model có quá tự tin khi sai không?
6. **Parameters + training time:** phần tăng chất lượng có đi cùng chi phí mô hình không?

Đây là cách biến phần "so sánh model" thành một báo cáo kỹ thuật thay vì chỉ nhìn accuracy.
