The target variable is `resale_price`, which is a continuous numerical value.

For example, a prediction may be:

```text
Predicted price = 428000
```

This is different from classification, where the output is a class label such as 0 or 1.

Our goal is therefore:

```text
Input features -> model -> predicted resale price
```

The main evaluation metrics will be:

- MAE: average absolute prediction error
- RMSE: square root of average squared error
- R2: proportion of variance explained by the model
- MAPE: average percentage error

For MAE, RMSE, and MAPE, lower values are better.
For R2, higher values are better.

# 6. Data Type Cleaning and Feature Engineering

Some useful information is stored as text and should be converted into numerical features before modeling.

### Feature decisions

1. `month`
   - Parse into a real date.
   - Extract transaction year and transaction month.

2. `remaining_lease`
   - Convert strings such as `61 years 04 months` into a numerical number of years.

3. `storey_range`
   - Convert ranges such as `10 TO 12` into the midpoint `11`.
   - Keep the original categorical storey range as a separate categorical feature.

4. `block` and `street_name`
   - They are high-cardinality categorical variables.
   - Direct one-hot encoding would create many columns.
   - We later use training-set frequency encoding for these two columns.

5. `resale_price`
   - This is the target and is never used as an input feature.

````python
def parse_remaining_lease(value):
    if pd.isna(value):
        return np.nan

    text = str(value).strip().lower()

    year_match = re.search(r"(\d+)\s+years?", text)
    month_match = re.search(r"(\d+)\s+months?", text)

    years = int(year_match.group(1)) if year_match else 0
    months = int(month_match.group(1)) if month_match else 0

    return years + months / 12.0


def parse_storey_mid(value):
    if pd.isna(value):
        return np.nan

    nums = re.findall(r"\d+", str(value))
    if len(nums) >= 2:
        return (int(nums[0]) + int(nums[1])) / 2.0
    if len(nums) == 1:
        return float(nums[0])

    return np.nan


df["month_date"] = pd.to_datetime(df["month"], format="%Y-%m", errors="coerce")
df["transaction_year"] = df["month_date"].dt.year
df["transaction_month"] = df["month_date"].dt.month
df["remaining_lease_years"] = df["remaining_lease"].apply(parse_remaining_lease)
df["storey_mid"] = df["storey_range"].apply(parse_storey_mid)

print("New engineered columns:")
print([
    "transaction_year",
    "transaction_month",
    "remaining_lease_years",
    "storey_mid"
])

display(
    df[
        [
            "month",
            "transaction_year",
            "transaction_month",
            "remaining_lease",
            "remaining_lease_years",
            "storey_range",
            "storey_mid"
        ]
    ].head()
)
````

### Invalid-value handling

Before modeling, we check logical constraints:

- `floor_area_sqm` should be greater than 0.
- `resale_price` should be greater than 0.
- parsed `remaining_lease_years` should not be negative.
- parsed `storey_mid` should not be negative.

Only records that violate a basic physical or mathematical constraint are removed.

We do **not** remove statistical outliers merely because they are large. Expensive flats are legitimate observations in a house-price problem, and automatically deleting them can remove meaningful market information.

````python
invalid_mask = (
    (df["floor_area_sqm"] <= 0)
    | (df["resale_price"] <= 0)
    | (df["remaining_lease_years"] < 0)
    | (df["storey_mid"] < 0)
)

print("Invalid logical rows:", int(invalid_mask.sum()))

df = df.loc[~invalid_mask].reset_index(drop=True)

print("Shape after logical validation:", df.shape)
````

````python
# Final missing-value audit after parsing engineered features.
missing_after_engineering = (
    df.isna()
    .sum()
    .sort_values(ascending=False)
    .to_frame("missing_count")
)

missing_after_engineering["missing_percent"] = (
    missing_after_engineering["missing_count"] / len(df) * 100
).round(3)

display(missing_after_engineering[missing_after_engineering["missing_count"] > 0])
````

The original dataset has no missing values, but parsing text fields can create missing values if an unexpected format is encountered.

The audit cell above checks this explicitly.

If any missing value appears:

- numeric engineered/input feature -> median imputation

# 8. Outlier Analysis

We use the IQR rule only as a diagnostic:

```text
IQR = Q3 - Q1
Lower bound = Q1 - 1.5 * IQR
Upper bound = Q3 + 1.5 * IQR
```

An observation outside these bounds is statistically unusual, but not automatically wrong.

For this dataset, we keep valid price and property observations even when they are statistical outliers. The reason is that the goal is to predict real resale prices, including legitimate expensive properties.

````python
def iqr_outlier_count(series):
    q1 = series.quantile(0.25)
    q3 = series.quantile(0.75)
    iqr = q3 - q1
    lower = q1 - 1.5 * iqr
    upper = q3 + 1.5 * iqr
    count = int(((series < lower) | (series > upper)).sum())
    return count, lower, upper

outlier_report = []

for col in [
    "floor_area_sqm",
    "remaining_lease_years",
    "storey_mid",
    "resale_price",
]:
    count, lower, upper = iqr_outlier_count(df[col].dropna())
    outlier_report.append({
        "feature": col,
        "iqr_outlier_count": count,
        "lower_bound": lower,
        "upper_bound": upper,
    })

display(pd.DataFrame(outlier_report))
````

# 9. Prepare the Modeling Table

We now separate:

- target: `resale_price`
- input features

The raw text column `month` is replaced by its parsed year and month.
The raw text column `remaining_lease` is replaced by the numerical `remaining_lease_years` representation.
The original `month_date` is also excluded from the model because it is only needed for EDA.

### Encoding strategy

Low-cardinality categorical features:

- `town`
- `flat_type`
- `storey_range`
- `flat_model`

These are one-hot encoded.

High-cardinality categorical features:

- `block`
- `street_name`

These use **frequency encoding**:

```text
frequency(category) = count(category) / number_of_training_rows
```

The frequency map is learned from the training split only. This prevents test-set information from influencing preprocessing.

````python
target_col = "resale_price"

categorical_low = [
    "town",
    "flat_type",
    "storey_range",
    "flat_model",
]

categorical_high = [
    "block",
    "street_name",
]

numeric_features = [
    "floor_area_sqm",
    "lease_commence_date",
    "transaction_year",
    "transaction_month",
    "remaining_lease_years",
    "storey_mid",
]

feature_columns = (
    categorical_low
    + categorical_high
    + numeric_features
)

X = df[feature_columns].copy()
y = df[target_col].copy()

print("X shape:", X.shape)
print("y shape:", y.shape)
display(X.head())
````

# 10. Train/Test Split

We use an 80/20 random split with a fixed seed so that:

- 80% of the observations are used for training.
- 20% are held out for final testing.
- the same split is used by all four models.
- the comparison is therefore fair.

The preprocessing statistics are learned from the training set only.

````python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=SEED,
)

print("Training samples:", len(X_train))
print("Testing samples :", len(X_test))
````

# 11. Leakage-Safe Preprocessing

This cell implements the preprocessing without relying on a black-box preprocessing pipeline, so that every step is visible.

### Numeric features

Missing values are filled with the training median.

### Categorical low-cardinality features

Missing values are filled with the training mode, then one-hot encoded.

### High-cardinality features

`block` and `street_name` are replaced with their training-set frequencies.

### Scaling

Only continuous numerical features are standardized:

```text
z = (x - mean) / std
```

The mean and standard deviation are calculated from `X_train` only.

The same training statistics are then applied to `X_test`.

This is exactly the same principle shown in the A3 Deep Learning example: the test set must use training-set normalization parameters.

````python
def fit_frequency_map(series):
    freq = series.value_counts(dropna=False) / len(series)
    return freq.to_dict()


def apply_frequency_encoding(series, mapping):
    # Unknown test categories receive frequency 0.
    return series.map(mapping).fillna(0.0)


def fit_transform_features(X_train, X_test):
    train = X_train.copy()
    test = X_test.copy()

    # 1. Numeric imputation using training medians.
    train_medians = train[numeric_features].median()

    train[numeric_features] = train[numeric_features].fillna(train_medians)
    test[numeric_features] = test[numeric_features].fillna(train_medians)

    # 2. High-cardinality categorical features -> frequency encoding.
    frequency_maps = {}

    for col in categorical_high:
        train[col] = train[col].astype(str)
        test[col] = test[col].astype(str)

        freq_map = fit_frequency_map(train[col])
        frequency_maps[col] = freq_map

        train[col + "_freq"] = apply_frequency_encoding(train[col], freq_map)
        test[col + "_freq"] = apply_frequency_encoding(test[col], freq_map)

    # 3. Low-cardinality categoricals -> training-based modes + one-hot encoding.
    train_modes = {}
    for col in categorical_low:
        mode = train[col].mode(dropna=True)
        mode_value = mode.iloc[0] if len(mode) else "UNKNOWN"
        train_modes[col] = mode_value
        train[col] = train[col].fillna(mode_value).astype(str)
        test[col] = test[col].fillna(mode_value).astype(str)

    train_dummies = pd.get_dummies(
        train[categorical_low],
        prefix=categorical_low,
        dtype=float
    )

    test_dummies = pd.get_dummies(
        test[categorical_low],
        prefix=categorical_low,
        dtype=float
    )

    # Make test columns exactly match training columns.
    test_dummies = test_dummies.reindex(
        columns=train_dummies.columns,
        fill_value=0.0
    )

    # 4. Build the continuous numerical part.
    continuous_cols = numeric_features + [
        col + "_freq" for col in categorical_high
    ]

    train_cont = pd.concat(
        [
            train[numeric_features].astype(float),
            train[[col + "_freq" for col in categorical_high]].astype(float),
        ],
        axis=1
    )

    test_cont = pd.concat(
        [
            test[numeric_features].astype(float),
            test[[col + "_freq" for col in categorical_high]].astype(float),
        ],
        axis=1
    )

    # 5. Standardize continuous columns using only training statistics.
    means = train_cont.mean()
    stds = train_cont.std().replace(0, 1.0)

    train_cont = (train_cont - means) / stds
    test_cont = (test_cont - means) / stds

    # 6. Combine continuous and one-hot features.
    train_final = pd.concat(
        [train_cont.reset_index(drop=True), train_dummies.reset_index(drop=True)],
        axis=1
    )

    test_final = pd.concat(
        [test_cont.reset_index(drop=True), test_dummies.reset_index(drop=True)],
        axis=1
    )

    preprocessing_info = {
        "numeric_medians": train_medians,
        "categorical_modes": train_modes,
        "frequency_maps": frequency_maps,
        "continuous_means": means,
        "continuous_stds": stds,
        "feature_names": train_final.columns.tolist(),
        "continuous_columns": continuous_cols,
        "dummy_columns": train_dummies.columns.tolist(),
    }

    return (
        train_final.astype(np.float32),
        test_final.astype(np.float32),
        preprocessing_info,
    )


X_train_processed, X_test_processed, preprocessing_info = fit_transform_features(
    X_train,
    X_test
)

print("Processed training shape:", X_train_processed.shape)
print("Processed testing shape :", X_test_processed.shape)
print("Number of final features:", X_train_processed.shape[1])
````

### Preprocessing verification

The processed matrices must satisfy:

- same number of columns
- no missing values
- no infinite values
- test data transformed using the training statistics

````python
Xtr = X_train_processed.to_numpy(dtype=np.float64)
Xte = X_test_processed.to_numpy(dtype=np.float64)

print("Same feature count:", Xtr.shape[1] == Xte.shape[1])
print("Training NaNs:", np.isnan(Xtr).sum())
print("Testing NaNs :", np.isnan(Xte).sum())
print("Training infs:", np.isinf(Xtr).sum())
print("Testing infs :", np.isinf(Xte).sum())
````

# 12. Model 1 - Linear Regression

Linear Regression provides a strong and interpretable baseline.

Conceptually:

```text
y_hat = XW + b
```

The model assumes that the target can be approximated as a weighted combination of the input features.

It is useful as a baseline because more complex models should demonstrate a meaningful improvement over this simpler mapping.

````python
y_train_np = y_train.to_numpy(dtype=np.float64)
y_test_np = y_test.to_numpy(dtype=np.float64)

model_lr = LinearRegression()

start = time.time()
model_lr.fit(Xtr, y_train_np)
lr_train_time = time.time() - start

pred_lr = model_lr.predict(Xte)

print("Linear Regression training time:", round(lr_train_time, 2), "seconds")
````

# 13. Model 2 - Random Forest Regressor

Random Forest builds many decision trees and averages their predictions.

Main idea:

```text
Input -> many decision trees -> average predictions
```

It can capture nonlinear relationships that a linear model cannot represent.

The model is trained on exactly the same processed training matrix and evaluated on exactly the same test matrix.

````python
model_rf = RandomForestRegressor(
    n_estimators=160,
    max_depth=20,
    min_samples_leaf=2,
    random_state=SEED,
    n_jobs=-1,
)

start = time.time()
model_rf.fit(Xtr, y_train_np)
rf_train_time = time.time() - start

pred_rf = model_rf.predict(Xte)

print("Random Forest training time:", round(rf_train_time, 2), "seconds")
````

# 14. Model 3 - HistGradientBoosting Regressor

HistGradientBoosting builds trees sequentially.

Each new tree focuses on reducing the errors made by the current ensemble.

Conceptually:

```text
Model 1 -> residual errors -> Model 2 -> residual errors -> ...
```

This is a boosting approach and is useful for comparing a sequential ensemble with Random Forest.

````python
model_hgb = HistGradientBoostingRegressor(
    max_iter=300,
    learning_rate=0.08,
    max_leaf_nodes=31,
    l2_regularization=0.1,
    random_state=SEED,
)

start = time.time()
model_hgb.fit(Xtr, y_train_np)
hgb_train_time = time.time() - start

pred_hgb = model_hgb.predict(Xte)

print(
    "HistGradientBoosting training time:",
    round(hgb_train_time, 2),
    "seconds"
)
````

# 15. Evaluate the Three Traditional ML Models

We now evaluate the models with the same metrics.

### MAE

```text
MAE = mean(abs(y - y_hat))
```

Interpretation: average absolute prediction error.

### RMSE

```text
RMSE = sqrt(mean((y - y_hat)^2))
```

RMSE penalizes large errors more strongly than MAE.

### R2

```text
R2 = 1 - residual_sum_of_squares / total_sum_of_squares
```

Higher is better.

### MAPE

```text
MAPE = mean(abs((y - y_hat) / y)) * 100
```

Lower is better.

MAPE must be interpreted carefully when true target values are near zero. In this dataset resale prices are positive and far from zero, so it is usable as an additional metric.

````python
def regression_metrics(y_true, y_pred):
    mae = mean_absolute_error(y_true, y_pred)
    rmse = np.sqrt(mean_squared_error(y_true, y_pred))
    r2 = r2_score(y_true, y_pred)

    y_true_arr = np.asarray(y_true, dtype=float)
    y_pred_arr = np.asarray(y_pred, dtype=float)
    mape = np.mean(
        np.abs((y_true_arr - y_pred_arr) / np.maximum(np.abs(y_true_arr), 1e-8))
    ) * 100

    return {
        "MAE": mae,
        "RMSE": rmse,
        "R2": r2,
        "MAPE (%)": mape,
    }


traditional_results = pd.DataFrame(
    [
        {"Model": "Linear Regression", **regression_metrics(y_test, pred_lr)},
        {"Model": "Random Forest", **regression_metrics(y_test, pred_rf)},
        {"Model": "HistGradientBoosting", **regression_metrics(y_test, pred_hgb)},
    ]
)

display(traditional_results.round(3))
````

# 17. Deep Learning from Scratch - Concept

The provided A3 material emphasizes the central idea:

```text
Deep Learning = composition of learned functions
```

For HousePrice, we use a regression network:

```text
Input features
    |
    v
Dense Layer 1: input_dim -> 32
    |
   ReLU
    |
    v
Dense Layer 2: 32 -> 16
    |
   ReLU
    |
    v
Output Layer: 16 -> 1
    |
   Linear output
    |
    v
Predicted price
```

This follows the same conceptual pattern as the A3 diabetes example:

```text
X -> H1 -> H2 -> y_hat
```

The important difference is the task:

- Diabetes classification -> sigmoid output + binary cross-entropy
- HousePrice regression -> linear output + mean squared error

# 18. Deep Learning Mathematics

For the first layer:

```text
Z1 = XW1 + b1
H1 = ReLU(Z1)
```

For the second layer:

```text
Z2 = H1W2 + b2
H2 = ReLU(Z2)
```

For the output:

```text
Z3 = H2W3 + b3
y_hat = Z3
```

Because this is regression, we do not apply sigmoid to the final output.

### ReLU

```text
ReLU(x) = max(0, x)
```

### Mean Squared Error

```text
MSE = (1/N) * sum((y_hat - y)^2)
```

### Learning objective

We want model parameters that minimize MSE.

The complete learning cycle remains:

```text
Forward -> Loss -> Backpropagation -> Gradient Descent
```

# 19. Prepare the Data for the NumPy Neural Network

The neural network below uses NumPy directly.

No sklearn model or sklearn neural-network implementation is used.

Because house prices are large values, the target is standardized for neural-network training:

```text
y_scaled = (y - mean_train) / std_train
```

After prediction, the values are transformed back to the original price scale.

This improves numerical stability while keeping the final MAE, RMSE, R2, and MAPE interpretable in real price units.

````python
X_train_dl = X_train_processed.to_numpy(dtype=np.float64)
X_test_dl = X_test_processed.to_numpy(dtype=np.float64)

y_train_raw = y_train_np.reshape(-1, 1)
y_test_raw = y_test.to_numpy(dtype=np.float64).reshape(-1, 1)

y_mean = y_train_raw.mean()
y_std = y_train_raw.std() + 1e-8

y_train_dl = (y_train_raw - y_mean) / y_std

print("X_train:", X_train_dl.shape)
print("X_test :", X_test_dl.shape)
print("y_train:", y_train_dl.shape)
print("Target mean used for scaling:", round(y_mean, 3))
print("Target std used for scaling :", round(y_std, 3))
````

# 20. Step 1 - Implement ReLU

This is directly analogous to the A3 example.

ReLU introduces nonlinearity, which allows the network to learn nonlinear relationships.

Its derivative is needed during backpropagation:

```text
ReLU'(x) = 0 if x <= 0
           1 if x > 0
```

````python
def relu(x):
    return np.maximum(0.0, x)


def relu_derivative(x):
    return (x > 0).astype(np.float64)
````

# 21. Step 2 - Initialize the Network

Architecture:

```text
input_dim -> 32 -> 16 -> 1
```

Weights start from small random values.

We use He-style initialization for the ReLU layers:

```text
W ~ Normal(0, 2 / fan_in)
```

Biases start at zero.

The dimensions are:

```text
W1: input_dim x 32
b1: 1 x 32

W2: 32 x 16
b2: 1 x 16

W3: 16 x 1
b3: 1 x 1
```

````python
input_dim = X_train_dl.shape[1]
hidden1 = 32
hidden2 = 16

rng = np.random.default_rng(SEED)

W1 = rng.normal(
    0,
    np.sqrt(2.0 / input_dim),
    size=(input_dim, hidden1)
)
b1 = np.zeros((1, hidden1))

W2 = rng.normal(
    0,
    np.sqrt(2.0 / hidden1),
    size=(hidden1, hidden2)
)
b2 = np.zeros((1, hidden2))

W3 = rng.normal(
    0,
    np.sqrt(2.0 / hidden2),
    size=(hidden2, 1)
)
b3 = np.zeros((1, 1))

print("W1:", W1.shape, "b1:", b1.shape)
print("W2:", W2.shape, "b2:", b2.shape)
print("W3:", W3.shape, "b3:", b3.shape)
````

# 22. Step 3 - Forward Propagation

The forward pass computes:

```text
Z1 = X @ W1 + b1
H1 = ReLU(Z1)

Z2 = H1 @ W2 + b2
H2 = ReLU(Z2)

Z3 = H2 @ W3 + b3
y_hat = Z3
```

The intermediate values are stored because backpropagation needs them later.

This is the same idea shown in the A3 tutorial, where the `cache` stores the intermediate tensors.

````python
def forward(X):
    z1 = X @ W1 + b1
    h1 = relu(z1)

    z2 = h1 @ W2 + b2
    h2 = relu(z2)

    z3 = h2 @ W3 + b3
    y_hat = z3

    cache = {
        "X": X,
        "z1": z1,
        "h1": h1,
        "z2": z2,
        "h2": h2,
        "z3": z3,
        "y_hat": y_hat,
    }

    return y_hat, cache
````

# 23. Step 4 - Mean Squared Error Loss

For HousePrice regression:

```text
MSE = mean((y_hat - y)^2)
```

A small MSE means the predictions are close to the true targets.

The loss function tells the network how wrong its current predictions are.

````python
def mse_loss(y, y_hat):
    return np.mean((y_hat - y) ** 2)
````

# 24. Step 5 - Backpropagation

For MSE:

```text
dZ3 = 2 * (y_hat - y) / N
```

Then:

```text
dW3 = H2.T @ dZ3
db3 = sum(dZ3)

dH2 = dZ3 @ W3.T
dZ2 = dH2 * ReLU'(Z2)

dW2 = H1.T @ dZ2
db2 = sum(dZ2)

dH1 = dZ2 @ W2.T
dZ1 = dH1 * ReLU'(Z1)

dW1 = X.T @ dZ1
db1 = sum(dZ1)
```

This is the same chain-rule idea demonstrated in the supplied A3 material.

Backpropagation is simply the systematic application of the chain rule through the composed functions.

````python
def backward(y, cache):
    Xc = cache["X"]
    z1 = cache["z1"]
    h1 = cache["h1"]
    z2 = cache["z2"]
    h2 = cache["h2"]
    y_hat = cache["y_hat"]

    n = len(Xc)

    # Output layer: linear activation + MSE
    dz3 = 2.0 * (y_hat - y) / n
    dW3 = h2.T @ dz3
    db3 = np.sum(dz3, axis=0, keepdims=True)

    # Hidden layer 2
    dh2 = dz3 @ W3.T
    dz2 = dh2 * relu_derivative(z2)
    dW2 = h1.T @ dz2
    db2 = np.sum(dz2, axis=0, keepdims=True)

    # Hidden layer 1
    dh1 = dz2 @ W2.T
    dz1 = dh1 * relu_derivative(z1)
    dW1 = Xc.T @ dz1
    db1 = np.sum(dz1, axis=0, keepdims=True)

    gradients = {
        "dW1": dW1,
        "db1": db1,
        "dW2": dW2,
        "db2": db2,
        "dW3": dW3,
        "db3": db3,
    }

    return gradients
````

# 25. Step 6 - Gradient Descent

The parameter update follows:

```text
W = W - learning_rate * gradient
b = b - learning_rate * gradient
```

The learning cycle is:

```text
Forward
    ->
Loss
    ->
Backward
    ->
Update
```

Because this dataset is much larger than the small diabetes example, we use mini-batches. The mathematical operations inside each mini-batch are the same as the full-batch version.

````python
def reset_network(seed=SEED):
    global W1, b1, W2, b2, W3, b3

    local_rng = np.random.default_rng(seed)

    W1 = local_rng.normal(
        0,
        np.sqrt(2.0 / input_dim),
        size=(input_dim, hidden1)
    )
    b1 = np.zeros((1, hidden1))

    W2 = local_rng.normal(
        0,
        np.sqrt(2.0 / hidden1),
        size=(hidden1, hidden2)
    )
    b2 = np.zeros((1, hidden2))

    W3 = local_rng.normal(
        0,
        np.sqrt(2.0 / hidden2),
        size=(hidden2, 1)
    )
    b3 = np.zeros((1, 1))


reset_network()
print("Network reset completed.")
````

# 26. Step 7 - Training Loop

Hyperparameters:

- learning rate = 0.01
- epochs = 40
- batch size = 4096

The loss is recorded after every epoch so that the training process can be visualized.

A decreasing loss curve indicates that the network is learning to reduce its training error.

````python
learning_rate = 0.01
epochs = 40
batch_size = 4096

loss_history = []

num_samples = len(X_train_dl)

training_start = time.time()

for epoch in range(epochs):
    indices = rng.permutation(num_samples)

    epoch_losses = []

    for start_idx in range(0, num_samples, batch_size):
        batch_idx = indices[start_idx:start_idx + batch_size]

        X_batch = X_train_dl[batch_idx]
        y_batch = y_train_dl[batch_idx]

        # Forward
        y_hat_batch, cache = forward(X_batch)

        # Loss
        loss = mse_loss(y_batch, y_hat_batch)
        epoch_losses.append(loss)

        # Backward
        gradients = backward(y_batch, cache)

        # Gradient descent
        W1 -= learning_rate * gradients["dW1"]
        b1 -= learning_rate * gradients["db1"]

        W2 -= learning_rate * gradients["dW2"]
        b2 -= learning_rate * gradients["db2"]

        W3 -= learning_rate * gradients["dW3"]
        b3 -= learning_rate * gradients["db3"]

    epoch_loss = float(np.mean(epoch_losses))
    loss_history.append(epoch_loss)

    if (epoch + 1) == 1 or (epoch + 1) % 5 == 0:
        print(
            f"Epoch {epoch + 1:02d}/{epochs} | "
            f"MSE = {epoch_loss:.6f}"
        )

dl_train_time = time.time() - training_start

print("Deep Learning training time:", round(dl_train_time, 2), "seconds")
````

````python
plt.figure(figsize=(10, 5))
plt.plot(range(1, epochs + 1), loss_history, marker="o", markersize=3)
plt.title("Deep Learning Training Loss")
plt.xlabel("Epoch")
plt.ylabel("Training MSE on scaled target")
plt.grid(alpha=0.25)
plt.tight_layout()
plt.show()
````

### Training curve interpretation

The plot above is analogous to the A3 diabetes challenge that stores `loss_history` and plots epoch versus loss.

Expected behavior:

- early epochs: larger error
- later epochs: lower error
- successful training: overall downward trend

A small amount of fluctuation is normal because mini-batches are shuffled each epoch.

# 36. Important Questions and Answers

### 1. Why is HousePrice a regression problem?

Because `resale_price` is a continuous numerical target.

### 2. Why are exact duplicates removed?

Exact duplicates repeat the same observation and can give that observation unintended extra weight.

### 3. How are missing values handled?

For this uploaded CSV, no missing values are present. The general preprocessing rule is median imputation for numeric features and mode imputation for categorical features, using training-set statistics only.

### 4. Why are statistical outliers kept?

A high-priced property can be a valid real transaction. Removing all IQR outliers could remove useful information from the target distribution.

### 5. Why is `remaining_lease` converted to a number?

The original text representation is not directly suitable for mathematical models. Converting it to years preserves the underlying numerical information.

### 6. Why is the test set transformed using training statistics?

Using test-set statistics would allow information from the test set to influence the model pipeline. That is a form of data leakage.

### 7. Why does the deep learning output layer not use sigmoid?

Sigmoid is suitable for binary classification probabilities. HousePrice is a regression problem, so the final output should be a continuous number.

### 8. Why is MSE used for the deep learning loss?

MSE is a standard differentiable loss for regression and has a simple derivative for manual backpropagation.

### 9. What does backpropagation calculate?

It calculates how the loss changes with respect to each trainable weight and bias.

### 10. What does gradient descent do?

It updates the parameters in the direction that reduces the loss.

### 11. Are the four models compared fairly?

Yes. They use the same cleaned data, the same train/test split, the same target, and the same evaluation metrics.

### 12. Which model should be selected?

The final selection should consider all reported metrics and the practical purpose of the task. A model should not be chosen by comparing only one metric.
