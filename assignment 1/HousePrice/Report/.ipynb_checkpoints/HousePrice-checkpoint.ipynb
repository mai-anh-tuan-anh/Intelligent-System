## 4. Address Feature Engineering

The original `Address` column contains detailed text.

For example:

`Đường Nguyễn Văn Khối, Phường 11, Gò Vấp, Hồ Chí Minh`

Instead of using the entire address as a high-cardinality text feature,
we extract two meaningful location features:

- District: Gò Vấp
- City/Province: Hồ Chí Minh

This allows the model to capture regional differences in house prices
without creating thousands of unique address categories.

```python
# Make a copy so the original Address column remains available
df["Address_Copy"] = df["Address"].astype(str)

# Split each address by comma
address_parts = df["Address_Copy"].str.split(",")

# Example:
# "Đường Nguyễn Văn Khối, Phường 11, Gò Vấp, Hồ Chí Minh"
#
# parts:
# [0] = Đường Nguyễn Văn Khối
# [1] = Phường 11
# [2] = Gò Vấp
# [3] = Hồ Chí Minh

# The last part is the City/Province
df["City_Province"] = (
    address_parts
    .str[-1]
    .str.strip()
)

# The second-last part is the District
df["District"] = (
    address_parts
    .str[-2]
    .str.strip()
)

# Display the result
df[
    [
        "Address",
        "District",
        "City_Province"
    ]
].head(20)
```

## 7. Data Cleaning and Missing Values

The dataset contains a large number of missing values.

Missing values cannot be passed directly to all machine learning
models, so they must be handled before training.

For numerical features, the median is used because it is less sensitive
to extreme values than the mean.

For categorical features, the most frequent category is used as a
simple and reproducible imputation strategy.

```python
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
from sklearn.preprocessing import OneHotEncoder

numeric_features = [
    "Area",
    "Frontage",
    "Access Road",
    "Floors",
    "Bedrooms",
    "Bathrooms"
]

categorical_features = [
    "District",
    "City_Province",
    "House direction",
    "Balcony direction",
    "Legal status",
    "Furniture state"
]

target = "Price"

X = df[
    numeric_features + categorical_features
]

y = df[target]

print("Number of features:", X.shape[1])
print("Number of samples:", X.shape[0])
```

```python
from sklearn.model_selection import train_test_split

# Separate the target from the features
X = df.drop(columns=["Price"])
y = df["Price"]

# Split the dataset into training and testing subsets
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,    # 20% reserved for testing
    random_state=42    # Reproducible split
)

print("Training samples:", len(X_train))
print("Testing samples:", len(X_test))
```

## 8. Preprocessing Strategy

The dataset contains both numerical and categorical features.

### Numerical features

Missing values are replaced with the median, then features are standardized.

### Categorical features

Missing values are replaced with the most frequent category,
then categories are converted into one-hot encoded numerical features.

The preprocessing is implemented inside a Scikit-learn Pipeline so that
the same preprocessing learned from the training data can be applied
to new data and eventually to the mobile application.

```python
# Numerical preprocessing:
# 1. Fill missing values with median
# 2. Standardize numerical features
numeric_pipeline = Pipeline([
    (
        "imputer",
        SimpleImputer(strategy="median")
    ),
    (
        "scaler",
        StandardScaler()
    )
])

# Categorical preprocessing:
# 1. Fill missing values with most frequent category
# 2. Convert categories into numerical one-hot vectors
categorical_pipeline = Pipeline([
    (
        "imputer",
        SimpleImputer(strategy="most_frequent")
    ),
    (
        "encoder",
        OneHotEncoder(
            handle_unknown="ignore"
        )
    )
])

# Combine both preprocessing pipelines
preprocessor = ColumnTransformer([
    (
        "numeric",
        numeric_pipeline,
        numeric_features
    ),
    (
        "categorical",
        categorical_pipeline,
        categorical_features
    )
])
```

```python
# ============================================================
# MODEL 1: LINEAR REGRESSION
# ============================================================

from sklearn.linear_model import LinearRegression

linear_model = Pipeline([
    (
        "preprocessor",
        preprocessor
    ),
    (
        "model",
        LinearRegression()
    )
])

# Train the model
linear_model.fit(
    X_train,
    y_train
)

print("Linear Regression trained.")
```

```python
# ============================================================
# MODEL 2: RIDGE REGRESSION
# ============================================================

from sklearn.linear_model import Ridge

ridge_model = Pipeline([
    (
        "preprocessor",
        preprocessor
    ),
    (
        "model",
        Ridge(alpha=1.0)
    )
])

ridge_model.fit(
    X_train,
    y_train
)

print("Ridge Regression trained.")
```

```python
# ============================================================
# MODEL 3: SUPPORT VECTOR REGRESSION
# ============================================================

from sklearn.svm import SVR

svr_model = Pipeline([
    (
        "preprocessor",
        preprocessor
    ),
    (
        "model",
        SVR(
            kernel="rbf",
            C=100,
            epsilon=0.1
        )
    )
])

svr_model.fit(
    X_train,
    y_train
)

print("SVR trained.")
```

```python
# ============================================================
# MODEL 4: RANDOM FOREST REGRESSOR
# ============================================================

from sklearn.ensemble import RandomForestRegressor

random_forest_model = Pipeline([
    (
        "preprocessor",
        preprocessor
    ),
    (
        "model",
        RandomForestRegressor(
            n_estimators=200,
            random_state=42,
            n_jobs=-1
        )
    )
])

random_forest_model.fit(
    X_train,
    y_train
)

print("Random Forest trained.")
```

```python
# ============================================================
# MODEL 5: GRADIENT BOOSTING REGRESSOR
# ============================================================

from sklearn.ensemble import GradientBoostingRegressor

gradient_boosting_model = Pipeline([
    (
        "preprocessor",
        preprocessor
    ),
    (
        "model",
        GradientBoostingRegressor(
            n_estimators=200,
            learning_rate=0.05,
            max_depth=3,
            random_state=42
        )
    )
])

gradient_boosting_model.fit(
    X_train,
    y_train
)

print("Gradient Boosting trained.")
```

## 9. Evaluation Metrics

Five regression metrics are used:

### MAE
Mean Absolute Error.

It represents the average absolute prediction error in billion VND.

Lower is better.

### MSE
Mean Squared Error.

Large errors are penalized more strongly.

Lower is better.

### RMSE
Root Mean Squared Error.

It has the same unit as the target price.

Lower is better.

### R²
Coefficient of determination.

It measures how much variation in house prices is explained by the model.

Higher is better.

### MAPE
Mean Absolute Percentage Error.

It measures the average prediction error as a percentage.

Lower is better.

```python
# ============================================================
# REGRESSION EVALUATION FUNCTION
# ============================================================

from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

def evaluate_regression_model(
    model,
    X_test,
    y_test
):
    """
    Evaluate a regression model using:
    MAE, MSE, RMSE, R² and MAPE.
    """

    # Predict prices for unseen test data
    predictions = model.predict(X_test)

    # Mean Absolute Error
    mae = mean_absolute_error(
        y_test,
        predictions
    )

    # Mean Squared Error
    mse = mean_squared_error(
        y_test,
        predictions
    )

    # Root Mean Squared Error
    rmse = np.sqrt(mse)

    # R-squared
    r2 = r2_score(
        y_test,
        predictions
    )

    # Avoid division by zero when calculating MAPE
    non_zero_mask = y_test != 0

    mape = (
        np.mean(
            np.abs(
                (
                    y_test[non_zero_mask]
                    -
                    predictions[non_zero_mask]
                )
                /
                y_test[non_zero_mask]
            )
        )
        * 100
    )

    return {
        "MAE": mae,
        "MSE": mse,
        "RMSE": rmse,
        "R2": r2,
        "MAPE": mape
    }
```

```python
# ============================================================
# EVALUATE ALL FIVE MODELS
# ============================================================

models = {
    "Linear Regression": linear_model,
    "Ridge Regression": ridge_model,
    "SVR": svr_model,
    "Random Forest": random_forest_model,
    "Gradient Boosting": gradient_boosting_model
}

results = []

for model_name, model in models.items():

    metrics = evaluate_regression_model(
        model,
        X_test,
        y_test
    )

    metrics["Model"] = model_name

    results.append(metrics)

results_df = pd.DataFrame(results)

# Put the model name first
results_df = results_df[
    [
        "Model",
        "MAE",
        "MSE",
        "RMSE",
        "R2",
        "MAPE"
    ]
]

results_df
```

## 10. Model Comparison

For MAE, MSE, RMSE and MAPE, smaller values indicate better performance.

For R², larger values indicate better performance.

Therefore, the best model should not be selected using only one metric.
The results of all five metrics should be considered together.

```python
# ============================================================
# FIND BEST MODEL FOR EACH METRIC
# ============================================================

print("Best MAE:")
print(
    results_df.loc[
        results_df["MAE"].idxmin()
    ]
)

print("\nBest MSE:")
print(
    results_df.loc[
        results_df["MSE"].idxmin()
    ]
)

print("\nBest RMSE:")
print(
    results_df.loc[
        results_df["RMSE"].idxmin()
    ]
)

print("\nBest R2:")
print(
    results_df.loc[
        results_df["R2"].idxmax()
    ]
)

print("\nBest MAPE:")
print(
    results_df.loc[
        results_df["MAPE"].idxmin()
    ]
)
```

## 11. Hyperparameter Experiment

### Question

Does changing the number of trees in Random Forest affect prediction performance?

We keep the dataset, preprocessing, training/test split and evaluation
procedure unchanged, and only change `n_estimators`.

```python
# ============================================================
# HYPERPARAMETER EXPERIMENT:
# RANDOM FOREST NUMBER OF TREES
# ============================================================
from sklearn.pipeline import Pipeline
rf_experiment_results = []

for n_trees in [50, 100, 200, 300]:

    model = Pipeline([
        (
            "preprocessor",
            preprocessor
        ),
        (
            "model",
            RandomForestRegressor(
                n_estimators=n_trees,
                random_state=42,
                n_jobs=-1
            )
        )
    ])

    # Train the model
    model.fit(
        X_train,
        y_train
    )

    # Evaluate the model
    metrics = evaluate_regression_model(
        model,
        X_test,
        y_test
    )

    metrics["n_estimators"] = n_trees

    rf_experiment_results.append(
        metrics
    )

rf_experiment_df = pd.DataFrame(
    rf_experiment_results
)

rf_experiment_df
```

```python
# ============================================================
# PLOT RANDOM FOREST HYPERPARAMETER EXPERIMENT
# ============================================================

plt.figure(figsize=(8, 5))

plt.plot(
    rf_experiment_df["n_estimators"],
    rf_experiment_df["RMSE"],
    marker="o"
)

plt.xlabel("Number of Trees")
plt.ylabel("RMSE")
plt.title(
    "Effect of Number of Trees on Random Forest RMSE"
)

plt.show()
```
